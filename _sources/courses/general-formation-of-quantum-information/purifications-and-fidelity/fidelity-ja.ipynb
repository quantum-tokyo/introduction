# 忠実度

このパートでは、量子状態の間の*忠実度*について議論します。忠実度は量子状態の類似度、つまり量子状態がどれだけ「重なっている」かを表す尺度です。

2つの量子状態ベクトルが与えられたとき、それらに対応する純粋状態の間の忠実度は、量子状態ベクトルの内積の絶対値に等しくなります。
これは類似度を測る基本的な方法を与えます。結果は$0$から$1$の間の値になり、値が大きいほど類似度が高いことを表します。
特に、この値は直交する状態では（定義により）0になり、グローバル位相を除いて等しい状態では$1$になります。

直感的に言えば、忠実度は、この基本的な類似度の尺度を量子状態ベクトルから密度行列へと拡張したものと見なせます。

## 忠実度の定義

まず忠実度の定義から始めるのが適切でしょう。
一見すると、次の定義は奇妙で不可解に見え、扱いにくそうに思えるかもしれません。
しかし、この定義によって定まる関数は、多くの興味深い性質と、いくつもの別の定式化を持つことがわかります。そのため、最初の印象よりもずっと扱いやすいものです。

<Figure title="定義">
  定義

  $\rho$と$\sigma$を、同じ系の量子状態を表す密度行列とします。
  $\rho$と$\sigma$の間の*忠実度*は、トレース（Tr）を使って次のように定義されます。

  $$
  \operatorname{F}(\rho,\sigma) = \operatorname{Tr}\sqrt{\sqrt{\rho} \sigma \sqrt{\rho}}.
  $$
</Figure>

<Figure title="注意">
  注意

  これは一般的な定義ですが、ここで定義した量の*2乗*を忠実度と定義し、ここで定義した量を*ルート忠実度*（root-fidelity）と呼ぶことも一般的です。
  どちらの定義が正しいとか間違っているということはなく、本質的には好みの問題です。
  とはいえ、どちらの定義が使われているのかを常に注意して理解するか、明確にする必要があります。
</Figure>

定義の式の意味を理解するために、まず$\sqrt{\rho} \sigma \sqrt{\rho}$が半正定値行列であることに注意しましょう。$M = \sqrt{\sigma}\sqrt{\rho}$とすると、

$$
\sqrt{\rho} \sigma \sqrt{\rho} = M^{\dagger} M
$$

となるからです。
すべての半正定値行列と同様に、この半正定値行列は一意な半正定値の平方根を持ち、そのトレースが忠実度です。

任意の正方行列$M$について、2つの半正定値行列$M^{\dagger} M$と$M M^{\dagger}$の固有値は常に同じであり、したがってこれらの行列の平方根についても同じことが成り立ちます。
$M = \sqrt{\sigma}\sqrt{\rho}$を選び、正方行列のトレースはその固有値の和であるという事実を使うと、次のようになります。

$$
\begin{aligned}
\operatorname{F}(\rho,\sigma)
& = \operatorname{Tr}\sqrt{\sqrt{\rho} \sigma \sqrt{\rho}}\\
& = \operatorname{Tr}\sqrt{M^{\dagger} M} = \operatorname{Tr}\sqrt{M M^{\dagger}}\\
& = \operatorname{Tr}\sqrt{\sqrt{\sigma} \rho \sqrt{\sigma}}\\
& = \operatorname{F}(\sigma,\rho).
\end{aligned}
$$

したがって、定義からはすぐにはわかりませんが、忠実度は2つの引数について対称です。

### トレースノルムによる忠実度の表現

忠実度を表す同等な方法として、次の式があります。

$$
\operatorname{F}(\rho,\sigma) = \bigl\|\sqrt{\sigma}\sqrt{\rho}\bigr\|_1.
$$

ここで現れているのは*トレースノルム*です。トレースノルムには、前のレッスンで状態識別の文脈ですでに出会いました。
（正方とは限らない）行列$M$のトレースノルムは、

$$
\| M \|_1 = \operatorname{Tr}\sqrt{M^{\dagger} M},
$$

と定義でき、この定義を行列$\sqrt{\sigma}\sqrt{\rho}$に適用すると、定義の式が得られます。

（正方）行列$M$のトレースノルムを表す別の方法として、次の式があります。

$$
\| M \|_1 = \max_{U\:\text{unitary}} \bigl\vert \operatorname{Tr}(M U) \bigr\vert.
$$

ここで最大値は、$M$と同じ行数と列数を持つすべての*ユニタリー*行列$U$にわたって取ります。
この式をいまの状況に適用すると、忠実度の別の表現が得られます。

$$
\operatorname{F}(\rho,\sigma)
= \max_{U\:\text{unitary}} \bigl\vert\operatorname{Tr}\bigl( \sqrt{\sigma}\sqrt{\rho}\, U\bigr) \bigr\vert
$$

### 純粋状態の忠実度

忠実度の定義について最後に述べておくことは、すべての純粋状態は（密度行列として）それ自身の平方根に等しいということです。そのため、一方または両方の状態が純粋であるときには、忠実度の式をかなり簡単にできます。
特に、2つの状態のうち一方が純粋であれば、次の式が得られます。

$$
\operatorname{F}\bigl( \vert\phi\rangle\langle\phi\vert, \sigma \bigr)
= \sqrt{\langle \phi\vert \sigma \vert \phi \rangle}
$$

両方の状態が純粋であれば、このセクションの最初で述べたように、式は対応する量子状態ベクトルの内積の絶対値に簡単になります。

$$
\operatorname{F}\bigl( \vert\phi\rangle\langle\phi\vert, \vert\psi\rangle\langle\psi\vert \bigr)
= \bigl\vert \langle \phi\vert \psi \rangle \bigr\vert
$$

## 忠実度の基本的な性質

忠実度には、多くの注目すべき性質といくつかの別の定式化があります。
ここでは、基本的な性質をいくつか、証明なしで挙げます。

1. 同じサイズの任意の2つの密度行列$\rho$と$\sigma$について、忠実度$\operatorname{F}(\rho,\sigma)$は0と1の間にあります。つまり、$0\leq \operatorname{F}(\rho,\sigma) \leq 1$です。$\operatorname{F}(\rho,\sigma)=0$となるのは$\rho$と$\sigma$の像が直交する（したがって誤りなく識別できる）場合に限り、$\operatorname{F}(\rho,\sigma)=1$となるのは$\rho = \sigma$の場合に限ります。
2. 忠実度は*乗法的*です。つまり、2つの積状態の間の忠実度は、個々の忠実度の積に等しくなります。
   $$
   \operatorname{F}(\rho_1\otimes\cdots\otimes\rho_m,\sigma_1\otimes\cdots\otimes\sigma_m)
   = \operatorname{F}(\rho_1,\sigma_1)\cdots \operatorname{F}(\rho_m,\sigma_m).
   $$
3. 状態の間の忠実度は、任意のチャネルの作用のもとで減少しません。つまり、$\rho$と$\sigma$が密度行列で、$\Phi$がこれら2つの状態を入力として受け取れるチャネルであるなら、必ず次の式が成り立ちます。
   $$
   \operatorname{F}(\rho,\sigma) \leq \operatorname{F}(\Phi(\rho),\Phi(\sigma)).
   $$
4. *フックス・ファン・デ・グラーフの不等式*は、忠実度とトレース距離の間の密接な（ただし厳密ではない）関係を与えます。任意の2つの状態$\rho$と$\sigma$について、次の式が成り立ちます。
   $$
   1 - \frac{1}{2}\|\rho - \sigma\|_1 \leq \operatorname{F}(\rho,\sigma)
   \leq \sqrt{1 - \frac{1}{4}\|\rho - \sigma\|_1^2}.
   $$

最後の性質は、図の形で表すことができます。

![トレース距離と忠実度の関係を表すプロット](https://quantum.cloud.ibm.com/learning/images/courses/general-formulation-of-quantum-information/purifications-and-fidelity/FvdG-plot.svg)

具体的には、同じ系の状態$\rho$と$\sigma$をどのように選んでも、$y$軸と$\operatorname{F}(\rho,\sigma)$で交わる水平線と、$x$軸と$\frac{1}{2}\|\rho-\sigma\|_1$で交わる垂直線は、下を直線$y = 1-x$、上を単位円で囲まれた灰色の領域の中で交わらなければなりません。
実用的な観点から最も興味深いのは、この図の灰色の領域の左上の隅です。
2つの状態の間の忠実度が1に近ければ、それらのトレース距離は0に近く、*その逆も*成り立ちます。

## 穏やかな測定の補題

次に、*穏やかな測定の補題*（gentle measurement lemma）として知られる、単純ですが重要な事実を見ていきます。この補題は、忠実度を非破壊測定と結びつけるものです。
これはときどき登場する非常に有用な補題です。また、一見ぎこちない忠実度の定義のおかげで、実はこの補題の証明がとても簡単になるという点でも注目に値します。

設定は次のとおりです。
$\mathsf{X}$を状態$\rho$にある系とし、$\{P_0,\ldots,P_{m-1}\}$を$\mathsf{X}$の一般測定を表す半正定値行列の集まりとします。
さらに、系$\mathsf{X}$が状態$\rho$にあるときにこの測定を行うと、測定結果のうちの1つが非常に高い確率で得られるとします。
具体的に、高い確率で得られる測定結果を$0$とし、特に、小さな正の実数$\varepsilon > 0$について

$$
\operatorname{Tr}(P_0 \rho) > 1 - \varepsilon
$$

であると仮定しましょう。

穏やかな測定の補題が述べているのは、これらの仮定のもとでは、ナイマークの定理によって$\{P_0,\ldots,P_{m-1}\}$から得られる非破壊測定は、高い確率で得られる測定結果$0$が観測された場合に、$\rho$にわずかな乱れしか引き起こさないということです。

より具体的には、この補題は、$\rho$と、測定結果が$0$であるという条件のもとで非破壊測定から得られる状態との間の忠実度の2乗が、$1-\varepsilon$より大きいことを述べています。

$$
\operatorname{F}\Biggl(\rho,\frac{\sqrt{P_0}\rho\sqrt{P_0}}{\operatorname{Tr}(P_0\rho)}\Biggr)^2 > 1-\varepsilon.
$$

これを証明するには、測定に関する基本的な事実が必要です。
測定行列$P_0, \ldots, P_{m-1}$は半正定値で、その和は単位行列になります。このことから、$P_0$の固有値はすべて$0$から$1$の間の実数であると結論できます。
これは、任意の単位ベクトル$\vert\psi\rangle$について、各$a\in\{0,\ldots,m-1\}$に対して値$\langle \psi \vert P_a \vert \psi \rangle$が非負の実数であること（各$P_a$が半正定値だから）と、これらの数の和が1になることから導かれます。

$$
\sum_{a = 0}^{m-1} \langle \psi \vert P_a \vert \psi \rangle
= \langle \psi \vert \Biggl(\sum_{a = 0}^{m-1}  P_a \Biggr) \vert \psi \rangle
= \langle \psi \vert \mathbb{I} \vert \psi \rangle = 1.
$$

したがって、$\langle \psi \vert P_0 \vert \psi \rangle$は常に$0$から$1$の間の実数です。$\vert\psi\rangle$は、関心のある固有値に対応する単位固有ベクトルとして特に選べるので、このことから$P_0$のすべての固有値は$0$から$1$の間の実数であることがわかります。

この観察から、すべての密度行列$\rho$について次の不等式が成り立つと結論できます。

$$
\operatorname{Tr}\bigl( \sqrt{P_0} \rho\bigr) \geq \operatorname{Tr}\bigl( P_0 \rho\bigr)
$$

より詳しく述べると、スペクトル分解

$$
P_0 = \sum_{k=0}^{n-1} \lambda_k \vert\psi_k\rangle\langle\psi_k\vert
$$

から出発して、各$k = 0,\ldots,n-1$について$\langle \psi_k \vert \rho \vert \psi_k \rangle$が非負の実数であり、$\sqrt{\lambda_k} \geq \lambda_k$であるという事実から、

$$
\operatorname{Tr}\bigl( \sqrt{P_0} \rho\bigr)
= \sum_{k = 0}^{n-1} \sqrt{\lambda_k} \langle \psi_k \vert \rho \vert \psi_k \rangle
\geq \sum_{k = 0}^{n-1} \lambda_k \langle \psi_k \vert \rho \vert \psi_k \rangle
= \operatorname{Tr}\bigl( P_0 \rho\bigr)
$$

が結論できます。（$0$から$1$の間の数は、2乗しても大きくなることはありません。）

では、忠実度を計算し、先ほどの不等式を使って、穏やかな測定の補題を証明しましょう。
まず、関心のある式を簡単にします。

$$
\begin{aligned}
\operatorname{F}\Biggl(\rho,\frac{\sqrt{P_0}\rho\sqrt{P_0}}{\operatorname{Tr}(P_0\rho)}\Biggr)
& = \operatorname{Tr}\sqrt{\frac{\sqrt{\rho}\sqrt{P_0}\rho\sqrt{P_0}\sqrt{\rho}}{
\operatorname{Tr}(P_0\rho)}}\\
& = \operatorname{Tr}\sqrt{\Biggl(\frac{\sqrt{\rho}\sqrt{P_0}\sqrt{\rho}}{
\sqrt{\operatorname{Tr}(P_0\rho)}}\Biggr)^2}\\
& = \operatorname{Tr}\Biggl(\frac{\sqrt{\rho}\sqrt{P_0}\sqrt{\rho}}{
\sqrt{\operatorname{Tr}(P_0\rho)}}\Biggr)\\
& = \frac{\operatorname{Tr}\bigl(\sqrt{P_0}\rho\bigr)}{\sqrt{\operatorname{Tr}(P_0\rho)}}
\end{aligned}
$$

これらはすべて等式であることに注意してください。この時点では、先ほどの不等式（やほかのどの不等式）も使っていないので、忠実度の厳密な表現が得られています。
ここで先ほどの不等式を使うと、

$$
\operatorname{F}\Biggl(\rho,\frac{\sqrt{P_0}\rho\sqrt{P_0}}{\operatorname{Tr}(P_0\rho)}\Biggr)
= \frac{\operatorname{Tr}\bigl(\sqrt{P_0}\rho\bigr)}{\sqrt{\operatorname{Tr}(P_0\rho)}}
\geq \frac{\operatorname{Tr}\bigl(P_0\rho\bigr)}{\sqrt{\operatorname{Tr}(P_0\rho)}}
= \sqrt{\operatorname{Tr}\bigl(P_0\rho\bigr)}
$$

が結論でき、したがって両辺を2乗すると、次のようになります。

$$
\operatorname{F}\Biggl(\rho,\frac{\sqrt{P_0}\rho\sqrt{P_0}}{\operatorname{Tr}(P_0\rho)}\Biggr)^2
\geq \operatorname{Tr}\bigl(P_0\rho\bigr) > 1-\varepsilon.
$$

## ウルマンの定理

このレッスンの締めくくりとして、*ウルマンの定理*を見ていきます。これは忠実度に関する基本的な事実であり、忠実度を純粋化の概念と結びつけるものです。
この定理が述べているのは、簡単に言えば、任意の2つの量子状態の間の忠実度は、それらの状態の2つの純粋化の間の内積（の絶対値）の*最大値*に等しいということです。

<Figure title="定理">
  定理
    
  ウルマンの定理: $\rho$と$\sigma$を系$\mathsf{X}$の状態を表す密度行列とし、$\mathsf{Y}$を、少なくとも$\mathsf{X}$と同じ数の古典状態を持つ系とします。$\rho$と$\sigma$の間の忠実度は、次の式で与えられます。

  $$
   \operatorname{F}(\rho,\sigma) = \max\bigl\{ \vert \langle \phi \vert \psi \rangle \vert \,:\, \operatorname{Tr}_{\mathsf{Y}}\bigl(\vert\phi\rangle\langle\phi\vert\bigr) = \rho,\; \operatorname{Tr}_{\mathsf{Y}}\bigl(\vert\psi\rangle\langle\psi\vert\bigr) = \sigma\bigr\},
  $$

  ここで最大値は、$(\mathsf{X},\mathsf{Y})$のすべての量子状態ベクトル$\vert\phi\rangle$と$\vert\psi\rangle$にわたって取ります。
</Figure>

この定理は純粋化のユニタリー同値性を使って証明できますが、完全に単純というわけではなく、途中で1つ工夫を使います。

まず、2つの密度行列$\rho$と$\sigma$のスペクトル分解を考えます。

$$
\begin{aligned}
\rho & = \sum_{a = 0}^{n-1} p_a \vert u_a\rangle\langle u_a\vert \\[2mm]
\sigma & = \sum_{b = 0}^{n-1} q_b \vert v_b\rangle\langle v_b\vert
\end{aligned}
$$

2つの集まり$\{\vert u_0 \rangle,\ldots,\vert u_{n-1}\rangle\}$と$\{\vert v_0 \rangle,\ldots,\vert v_{n-1}\rangle\}$は、それぞれ$\rho$と$\sigma$の固有ベクトルからなる正規直交基底であり、$p_0,\ldots,p_{n-1}$と$q_0,\ldots,q_{n-1}$は対応する固有値です。

また、$\vert u_0 \rangle,\ldots,\vert u_{n-1}\rangle$と$\vert v_0 \rangle,\ldots,\vert v_{n-1}\rangle$の各成分の複素共役を取って得られるベクトルを、$\vert \overline{u_0} \rangle,\ldots,\vert \overline{u_{n-1}}\rangle$と$\vert \overline{v_0} \rangle,\ldots,\vert \overline{v_{n-1}}\rangle$と定義します。
つまり、任意のベクトル$\vert w\rangle$について、各$c\in\{0,\ldots,n-1\}$に対する次の等式によって$\vert\overline{w}\rangle$を定義します。

$$
\langle c \vert \overline{w}\rangle = \overline{\langle c \vert w\rangle}
$$

任意の2つのベクトル$\vert u\rangle$と$\vert v\rangle$について$\langle \overline{u} \vert \overline{v}\rangle = \langle v\vert u\rangle$が成り立つことに注意してください。
より一般に、任意の正方行列$M$について次の式が成り立ちます。

$$
\langle \overline{u} \vert M \vert \overline{v}\rangle = \langle v\vert M^T \vert u\rangle
$$

このことから、$\vert u\rangle$と$\vert v\rangle$が直交するのは$\vert \overline{u}\rangle$と$\vert \overline{v}\rangle$が直交する場合に限ることがわかります。したがって、$\{\vert \overline{u_0} \rangle,\ldots,\vert \overline{u_{n-1}}\rangle\}$と$\{\vert \overline{v_0} \rangle,\ldots,\vert \overline{v_{n-1}}\rangle\}$はどちらも正規直交基底です。

ここで、次の2つのベクトル$\vert\phi\rangle$と$\vert\psi\rangle$を考えます。これらはそれぞれ$\rho$と$\sigma$の純粋化です。

$$
\begin{aligned}
\vert\phi\rangle & = \sum_{a = 0}^{n-1} \sqrt{p_a}\, \vert u_a\rangle \otimes \vert \overline{u_a}\rangle \\[2mm]
\vert\psi\rangle & = \sum_{b = 0}^{n-1} \sqrt{q_b}\, \vert v_b\rangle \otimes \vert
\overline{v_b}\rangle
\end{aligned}
$$

これが先ほど述べた工夫です。
この時点では、$\rho$と$\sigma$の純粋化としてこれらを選ぶのがよい考えだと明示的に示すものは何もありません。しかし、これらは正しい純粋化であり、複素共役によって、必要なとおりに計算がうまく進むようになります。

純粋化のユニタリー同値性により、系の組$(\mathsf{X},\mathsf{Y})$に対する$\rho$のすべての純粋化は、あるユニタリー行列$U$を使って$(\mathbb{I}_{\mathsf{X}}\otimes U)\vert\phi\rangle$の形で表されなければならないことがわかります。同様に、組$(\mathsf{X},\mathsf{Y})$に対する$\sigma$のすべての純粋化は、あるユニタリー行列$V$を使って$(\mathbb{I}_{\mathsf{X}}\otimes V)\vert\psi\rangle$の形で表されなければなりません。
このような2つの純粋化の内積は、次のように簡単にできます。

$$
\begin{aligned}
\langle \phi \vert (\mathbb{I}\otimes U^{\dagger}) (\mathbb{I}\otimes V) \vert \psi \rangle
\hspace{-2.5cm}\\
& = \sum_{a,b = 0}^{n-1} \sqrt{p_a} \sqrt{q_b}\, \langle u_a \vert v_b \rangle
\langle \overline{u_a} \vert U^{\dagger} V \vert \overline{v_b} \rangle \\
& = \sum_{a,b = 0}^{n-1} \sqrt{p_a} \sqrt{q_b}\, \langle u_a \vert v_b \rangle
\langle v_b \vert (U^{\dagger} V)^T \vert u_a \rangle \\
& = \operatorname{Tr}\Biggl(
\sum_{a,b = 0}^{n-1} \sqrt{p_a} \sqrt{q_b}\, \vert u_a \rangle\langle u_a \vert v_b \rangle
\langle v_b \vert (U^{\dagger} V)^T\Biggr)\\
& = \operatorname{Tr}\Bigl(
\sqrt{\rho}\sqrt{\sigma}\, (U^{\dagger} V)^T\Bigr)
\end{aligned}
$$

$U$と$V$がすべてのユニタリー行列にわたって動くとき、行列$(U^{\dagger} V)^T$もすべてのユニタリー行列にわたって動きます。
したがって、$\rho$と$\sigma$の2つの純粋化の内積の絶対値を最大化すると、次の等式が得られます。

$$
\begin{aligned}
\max_{U,V\:\text{unitary}} \biggl\vert \operatorname{Tr}\Bigl(
\sqrt{\rho}\sqrt{\sigma}\, (U^{\dagger} V)^T\Bigr)\biggr\vert
& = \max_{W\:\text{unitary}} \biggl\vert \operatorname{Tr}\Bigl(
\sqrt{\rho}\sqrt{\sigma}\, W\Bigr)\biggr\vert\\[2mm]
& = \bigl\| \sqrt{\rho}\sqrt{\sigma} \bigr\|_1\\[2mm]
& = \operatorname{F}(\rho,\sigma)
\end{aligned}
$$
