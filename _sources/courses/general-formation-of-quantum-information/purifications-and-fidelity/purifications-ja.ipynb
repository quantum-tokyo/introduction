# 純粋化

## 純粋化の定義

まず、純粋化の正確な数学的定義から始めましょう。

<Figure title="定義">
  定義
    
  系$\mathsf{X}$が密度行列$\rho$で表される状態にあり、$\vert\psi\rangle$は組$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルで、$\mathsf{Y}$をトレースアウトすると$\rho$が残るものとします。

  $$
  \rho = \operatorname{Tr}_{\mathsf{Y}} \bigl( \vert \psi\rangle\langle\psi\vert\bigr).
  $$

  このとき、状態ベクトル$\vert\psi\rangle$を$\rho$の*純粋化*といいます。
</Figure>

定義の等式が成り立つとき、量子状態ベクトルではなく密度行列として表した純粋状態$\vert\psi\rangle\langle\psi\vert$も、一般に$\rho$の純粋化と呼ばれます。ただし、このレッスンでは、通常この用語を量子状態ベクトルを指すものとして使います。

*純粋化*という用語は、より一般に、系の順番が逆の場合や、（当然ながら）系や状態の名前が異なる場合、さらに系が3つ以上ある場合にも使われます。
たとえば、$\vert \psi \rangle$が複合系$(\mathsf{A},\mathsf{B},\mathsf{C})$の純粋状態を表す量子状態ベクトルで、系$(\mathsf{A},\mathsf{C})$の状態を表す密度行列$\rho$について等式

$$
\rho = \operatorname{Tr}_{\mathsf{B}} \bigl(\vert\psi\rangle\langle\psi\vert\bigr)
$$

が成り立つなら、$\vert\psi\rangle$はやはり$\rho$の純粋化と呼ばれます。

しかし、このレッスンでは、定義で述べた特定の形に焦点を当てます。
この定義に基づく純粋化の性質や事実は、通常、系を並べ替えて2つの複合系に分け、一方に$\mathsf{X}$の役割を、もう一方に$\mathsf{Y}$の役割を担わせることで、3つ以上の系の場合に一般化できます。

## 純粋化の存在

$\mathsf{X}$と$\mathsf{Y}$を任意の2つの系とし、$\rho$を$\mathsf{X}$の与えられた状態とします。
系$\mathsf{Y}$が十分に大きければ、$\rho$を*純粋化する*（これは$\vert\psi\rangle$が$\rho$の純粋化であることの別の言い方です）$(\mathsf{X},\mathsf{Y})$の量子状態ベクトル$\vert\psi\rangle$が存在することを証明します。
特に、$\mathsf{Y}$が少なくとも$\mathsf{X}$と同じ数の古典状態を持つなら、この形の純粋化はすべての状態$\rho$について必ず存在します。
状態$\rho$によっては、$\mathsf{Y}$の古典状態はもっと少なくて済みます。
一般に、$\rho$を純粋化する$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルが存在するためには、$\mathsf{Y}$の古典状態が$\operatorname{rank}(\rho)$個あることが必要十分です。

まず、任意の正の整数$n$について、$\rho$を$n$個の純粋状態の凸結合として表した任意の式を考えます。

$$
\rho = \sum_{a = 0}^{n-1} p_a \vert\phi_a\rangle\langle\phi_a\vert
$$

この式で、$(p_0,\ldots,p_{n-1})$は確率ベクトルであり、$\vert\phi_0\rangle,\ldots,\vert\phi_{n-1}\rangle$は$\mathsf{X}$の量子状態ベクトルです。

このような式を得る方法の1つはスペクトル定理を使うことです。その場合、$n$は$\mathsf{X}$の古典状態の数、$p_0,\ldots,p_{n-1}$は$\rho$の固有値、$\vert\phi_0\rangle,\ldots,\vert\phi_{n-1}\rangle$はこれらの固有値に対応する正規直交な固有ベクトルです。

実は、和の中に$\rho$の固有値0に対応する項を含める必要はありません。そのため、代わりに$n = \operatorname{rank}(\rho)$とし、$p_0,\ldots,p_{n-1}$を$\rho$の0でない固有値とすることもできます。
これは、上の形の$\rho$の式が存在する$n$の最小値です。

念のため述べておくと、純粋状態の凸結合として選ぶ$\rho$の式が、スペクトル定理から得られたものである*必要はありません*。スペクトル定理は、このような式を得る方法の1つにすぎません。
特に、$n$は任意の正の整数でよく、単位ベクトル$\vert\phi_0\rangle,\ldots,\vert\phi_{n-1}\rangle$は直交している必要がなく、確率$p_0,\ldots,p_{n-1}$は$\rho$の固有値である必要もありません。

これで、$\rho$の純粋化を次のように与えることができます。

$$
\vert\psi\rangle = \sum_{a = 0}^{n-1} \sqrt{p_a} \, \vert\phi_a\rangle \otimes \vert a \rangle
$$

ここでは、$\mathsf{Y}$の古典状態に$0,\ldots,n-1$が含まれると仮定しています。
含まれない場合は、$\mathsf{Y}$の互いに異なる$n$個の古典状態を任意に選んで、$0,\ldots,n-1$の代わりに使えます。
これが確かに$\rho$の純粋化であることは、部分トレースを計算するだけで簡単に確かめられます。部分トレースは、次の2つの同等な方法で計算できます。

$$
\operatorname{Tr}_{\mathsf{Y}} \bigl(\vert\psi\rangle\langle\psi\vert\bigr) =
\sum_{a = 0}^{n-1} (\mathbb{I}_{\mathsf{X}} \otimes \langle a\vert) \vert\psi\rangle\langle\psi\vert
(\mathbb{I}_{\mathsf{X}} \otimes \vert a\rangle) = \sum_{a = 0}^{n-1} p_a \vert\phi_a\rangle\langle\phi_a\vert = \rho
$$

$$
\operatorname{Tr}_{\mathsf{Y}} \bigl(\vert\psi\rangle\langle\psi\vert\bigr) =
\sum_{a,b = 0}^{n-1} \sqrt{p_a} \sqrt{p_b} \, \vert\phi_a\rangle\langle \phi_b\vert
\, \operatorname{Tr}(\vert a \rangle \langle b \vert)
= \sum_{a = 0}^{n-1} p_a \, \vert\phi_a\rangle\langle \phi_a\vert = \rho
$$

より一般に、任意の正規直交なベクトルの集合$\{\vert\gamma_0\rangle,\ldots,\vert\gamma_{n-1}\rangle\}$について、量子状態ベクトル

$$
\vert\psi\rangle = \sum_{a = 0}^{n-1} \sqrt{p_a} \, \vert\phi_a\rangle \otimes \vert \gamma_a \rangle
$$

は$\rho$の純粋化です。

### 例

$\mathsf{X}$と$\mathsf{Y}$がどちらも量子ビットであり、

$$
\rho = \begin{pmatrix}
\frac{3}{4} & \frac{1}{4}\\[2mm]
\frac{1}{4} & \frac{1}{4}
\end{pmatrix}
$$

が$\mathsf{X}$の状態を表す密度行列であるとします。

スペクトル定理を使うと、$\rho$は次のように表せます。

$$
\rho =
\cos^2(\pi/8) \vert \psi_{\pi/8}\rangle\langle\psi_{\pi/8}\vert +
\sin^2(\pi/8) \vert \psi_{5\pi/8}\rangle\langle\psi_{5\pi/8}\vert,
$$

ここで$\vert \psi_{\theta} \rangle = \cos(\theta) \vert 0\rangle + \sin(\theta)\vert 1\rangle$です。
したがって、組$(\mathsf{X},\mathsf{Y})$の純粋状態を表す量子状態ベクトル

$$
\cos(\pi/8) \vert \psi_{\pi/8}\rangle \otimes \vert 0\rangle +
\sin(\pi/8) \vert \psi_{5\pi/8}\rangle \otimes \vert 1\rangle
$$

は、$\rho$の純粋化です。

あるいは、次のように書くこともできます。

$$
\rho = \frac{1}{2} \vert 0\rangle\langle 0\vert + \frac{1}{2} \vert +\rangle\langle +\vert.
$$

これは純粋状態の凸結合ですが、スペクトル分解ではありません。$\vert 0\rangle$と$\vert +\rangle$は直交しておらず、$1/2$は$\rho$の固有値ではないからです。
それでも、量子状態ベクトル

$$
\frac{1}{\sqrt{2}} \vert 0 \rangle \otimes \vert 0\rangle +
\frac{1}{\sqrt{2}} \vert + \rangle \otimes \vert 1\rangle
$$

は$\rho$の純粋化です。

## シュミット分解

次に、*シュミット分解*について議論します。シュミット分解とは、系の*組*の量子状態ベクトルをある特定の形で表したものです。
シュミット分解は純粋化と密接に関係しており、それ自体としても非常に有用です。
実際、系の組の量子状態ベクトル$\vert\psi\rangle$について推論するとき、最初のステップはその状態のシュミット分解を求めるか考えることである場合が多いです。

<Figure title="定義">
  $\vert \psi\rangle$を系の組$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルとします。$\vert\psi\rangle$の*シュミット分解*とは、次の形の式のことです。

  $$
  \vert \psi\rangle = \sum_{a = 0}^{r-1} \sqrt{p_a}\, \vert x_a\rangle \otimes \vert y_a \rangle,
  $$

  ここで$p_0,\ldots,p_{r-1}$は和が$1$になる正の実数であり、集合$\{\vert x_0\rangle,\ldots,\vert x_{r-1}\rangle\}$と$\{\vert y_0\rangle,\ldots,\vert y_{r-1}\rangle\}$は*どちらも*正規直交です。
</Figure>

$\vert\psi\rangle$のシュミット分解に現れる値

$$
\sqrt{p_0},\ldots,\sqrt{p_{r-1}}
$$

はその*シュミット係数*と呼ばれ、（並び順を除いて）一意に決まります。つまり、$\vert\psi\rangle$をこのような形で表したときに現れうる正の実数は、これらだけです。
一方、集合

$$
\{\vert x_0\rangle,\ldots,\vert x_{r-1}\rangle\} \quad\text{and}\quad
\{\vert y_0\rangle,\ldots,\vert y_{r-1}\rangle\},
$$

は一意には決まりません。これらのベクトルの集合をどの程度自由に選べるかは、この後の説明で明らかになります。

では、与えられた量子状態ベクトル$\vert\psi\rangle$が確かにシュミット分解を持つことを確かめ、その過程でシュミット分解の求め方を学びましょう。

まず、系$\mathsf{X}$に対応するベクトル空間の任意の（直交とは限らない）基底$\{\vert x_0\rangle, \ldots, \vert x_{n-1}\rangle\}$を考えます。
これは基底なので、次の等式が成り立つようなベクトル$\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle$の選び方が、常にただ1通り存在します。

$$
\vert \psi\rangle = \sum_{a = 0}^{n-1} \vert x_a\rangle \otimes \vert z_a \rangle
\tag{1}
$$

たとえば、$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$が$\mathsf{X}$に対応する標準基底であるとします。
$\mathsf{X}$の古典状態集合を$\{0,\ldots,n-1\}$とすると、これは各$a\in\{0,\ldots,n-1\}$について$\vert x_a\rangle = \vert a\rangle$であることを意味します。このとき、

$$
\vert\psi\rangle = \sum_{a = 0}^{n-1} \vert a\rangle \otimes \vert z_a\rangle
$$

が成り立ちます。ここで、各$a\in\{0,\ldots,n-1\}$について

$$
\vert z_a \rangle = ( \langle a \vert \otimes \mathbb{I}_{\mathsf{Y}}) \vert \psi\rangle
$$

です。
$\mathsf{X}$の標準基底測定を考えるときには、このような式をよく扱います。

次の点に注意することが重要です。この例でベクトル$\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle$を与える式

$$
\vert z_a \rangle = ( \langle a \vert \otimes \mathbb{I}_{\mathsf{Y}}) \vert \psi\rangle
$$

が成り立つのは、$\{\vert 0\rangle,\ldots,\vert n-1\rangle\}$が*正規直交*基底だからにすぎません。
一般に、$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$が正規直交とは限らない基底の場合でも、ベクトル$\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle$は等式$(1)$によって一意に決まりますが、別の式が必要になります。
それらを求める方法の1つは、まずすべての$a,b\in\{0,\ldots,n-1\}$について等式

$$
\langle w_a \vert x_b \rangle = \begin{cases} 1 & a=b\\ 0 & a\neq b \end{cases}
$$

を満たすベクトル$\vert w_0\rangle,\ldots,\vert w_{n-1}\rangle$を求めることです。すると、

$$
\vert z_a \rangle = (\langle w_a \vert \otimes \mathbb{I}_{\mathsf{Y}}) \vert \psi\rangle.
$$

となります。

$\mathsf{X}$に対応するベクトル空間の基底$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$が与えられたとき、等式$(1)$を満たす一意に決まるベクトル$\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle$は、$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$がたまたま正規直交基底であったとしても、特別な性質を満たすとは限りません。
しかし、$\{\vert x_0\rangle, \ldots, \vert x_{n-1}\rangle\}$を縮約状態

$$
\rho = \operatorname{Tr}_{\mathsf{Y}} \bigl( \vert \psi\rangle \langle \psi \vert \bigr),
$$

の*固有ベクトル*からなる正規直交基底に選ぶと、興味深いことが起こります。
具体的には、等式$(1)$を成り立たせる一意に決まる集まり$\{\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle\}$は、必ず*直交*します。

より詳しく、$\rho$のスペクトル分解を考えます。

$$
\rho = \sum_{a = 0}^{n-1} p_a \vert x_a \rangle \langle x_a \vert
$$

ここで、$\rho$が密度行列である（したがって固有値のベクトル$(p_0,\ldots,p_{n-1})$は確率ベクトルになる）ことを踏まえて、$\rho$の固有値を$p_0,\ldots,p_{n-1}$で表しています。また、$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$はこれらの固有値に対応する固有ベクトルからなる正規直交基底です。
等式$(1)$を成り立たせる一意な集まり$\{\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle\}$が必ず直交することを確かめるには、まず部分トレースを計算します。

$$
\begin{aligned}
\operatorname{Tr}_{\mathsf{Y}} (\vert\psi\rangle\langle\psi\vert)
& = \sum_{a,b = 0}^{n-1} \vert x_a\rangle\langle x_b\vert \operatorname{Tr}(\vert z_a\rangle\langle z_b\vert)\\
& = \sum_{a,b = 0}^{n-1} \langle z_b\vert z_a\rangle \, \vert x_a\rangle\langle x_b\vert.
\end{aligned}
$$

この式は、$\rho$のスペクトル分解と一致しなければなりません。
$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$は基底なので、行列の集合

$$
\bigl\{ \vert x_a\rangle\langle x_b\vert \,:\, a,b\in\{0,\ldots,n-1\} \bigr\}
$$

は線形独立であり、したがって

$$
\langle z_b \vert z_a\rangle =
\begin{cases}
  p_a & a=b\\[1mm]
  0 & a\neq b,
\end{cases}
$$

が成り立ちます。これで、$\{\vert z_0\rangle,\ldots,\vert z_{n-1}\rangle\}$が直交することが示されました。

これで$\vert\psi\rangle$のシュミット分解がほぼ得られました。
残りは、$(1)$のうち$p_a = 0$となる項を取り除き、残りの各項について、単位ベクトル$\vert y_a\rangle$を使って$\vert z_a\rangle = \sqrt{p_a}\vert y_a\rangle$と書くことです。

これを行う便利な方法は、縮約状態$\rho$のスペクトル分解における固有値と固有ベクトルの組には好きなように番号を付けてよい、という点に注目することから始まります。そこで、固有値は降順に並んでいると仮定できます。

$$
p_0 \geq p_1 \geq \cdots \geq p_{n-1}.
$$

$r = \operatorname{rank}(\rho)$とすると、$p_0,\ldots,p_{r-1} > 0$かつ$p_r = \cdots = p_{n-1} = 0$となります。
したがって、

$$
\rho = \sum_{a = 0}^{r-1} p_a \vert x_a \rangle \langle x_a \vert,
$$

であり、量子状態ベクトル$\vert \psi \rangle$は次のように書けます。

$$
\vert\psi\rangle = \sum_{a = 0}^{r-1} \vert x_a\rangle \otimes \vert z_a\rangle.
$$

$a=0,\ldots,r-1$について

$$
\| \vert z_a \rangle \|^2 = \langle z_a \vert z_a \rangle = p_a > 0
$$

であることから、単位ベクトル$\vert y_0 \rangle,\ldots,\vert y_{r-1}\rangle$を

$$
\vert y_a\rangle = \frac{\vert z_a\rangle}{\|\vert z_a\rangle\|} = \frac{\vert z_a\rangle}{\sqrt{p_a}},
$$

と定義できます。すると、各$a\in\{0,\ldots,r-1\}$について$\vert z_a\rangle = \sqrt{p_a}\vert y_a\rangle$となります。
ベクトル$\{\vert z_0\rangle, \ldots, \vert z_{r-1}\rangle\}$は直交していて0でないので、$\{\vert y_0\rangle, \ldots, \vert y_{r-1}\rangle\}$は*正規直交*集合になります。こうして、$\vert\psi\rangle$のシュミット分解が得られました。

$$
\vert \psi\rangle = \sum_{a = 0}^{r-1} \sqrt{p_a}\, \vert x_a\rangle \otimes \vert y_a \rangle
$$

ベクトル$\{\vert x_0\rangle,\ldots,\vert x_{r-1}\rangle\}$と$\{\vert y_0\rangle,\ldots,\vert y_{r-1}\rangle\}$の選び方については、（上で行ったように）$\{\vert x_0\rangle,\ldots,\vert x_{r-1}\rangle\}$を、縮約状態$\operatorname{Tr}_{\mathsf{Y}}(\vert\psi\rangle\langle\psi\vert)$の0でない固有値に対応する固有ベクトルからなる任意の正規直交集合に選べます。その場合、ベクトル$\{\vert y_0\rangle,\ldots,\vert y_{r-1}\rangle\}$は一意に決まります。

状況は2つの系の間で対称なので、代わりに$\{\vert y_0\rangle,\ldots,\vert y_{r-1}\rangle\}$を、縮約状態$\operatorname{Tr}_{\mathsf{X}}(\vert\psi\rangle\langle\psi\vert)$の0でない固有値に対応する固有ベクトルからなる任意の正規直交集合に選ぶこともできます。その場合は、ベクトル$\{\vert x_0\rangle,\ldots,\vert x_{r-1}\rangle\}$が一意に決まります。

ただし、いま述べたように一方の集合を対応する縮約状態の固有ベクトルの集合として選ぶと、もう一方は決まってしまうことに注意してください。つまり、2つを独立に選ぶことはできません。

このシリーズでは再び登場しませんが、系の組$(\mathsf{X},\mathsf{Y})$の任意の純粋状態$\vert\psi\rangle$について、縮約状態$\operatorname{Tr}_{\mathsf{X}}(\vert\psi\rangle\langle\psi\vert)$の0でない固有値$p_0,\ldots,p_{r-1}$は、縮約状態$\operatorname{Tr}_{\mathsf{Y}}(\vert\psi\rangle\langle\psi\vert)$の0でない固有値と常に一致するという点は注目に値します。

直感的に言えば、組$(\mathsf{X},\mathsf{Y})$が純粋状態にあるとき、$\mathsf{X}$と$\mathsf{Y}$の縮約状態には、まったく同じ量のランダムさが含まれています。
この事実はシュミット分解から明らかになります。どちらの場合も、縮約状態の固有値は純粋状態のシュミット係数の2乗と一致しなければなりません。

## 純粋化のユニタリー同値性

シュミット分解を使うと、*純粋化のユニタリー同値性*として知られる、純粋化に関する根本的に重要な事実を示すことができます。

<Figure title="定理">
  純粋化のユニタリー同値性: $\mathsf{X}$と$\mathsf{Y}$を系とし、$\vert\psi\rangle$と$\vert\phi\rangle$を、$\mathsf{X}$の同じ状態をどちらも純粋化する$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルとします。記号で書けば、$\mathsf{X}$の状態を表すある密度行列$\rho$について

  $$
  \operatorname{Tr}_{\mathsf{Y}} (\vert\psi\rangle\langle\psi\vert) = \rho = \operatorname{Tr}_{\mathsf{Y}} (\vert\phi\rangle\langle\phi\vert)
  $$

  が成り立つということです。
  このとき、1つ目の純粋化を2つ目の純粋化に変換する、$\mathsf{Y}$だけに作用するユニタリー演算$U$が存在しなければなりません。

  $$
  (\mathbb{I}_{\mathsf{X}} \otimes U) \vert\psi\rangle = \vert\phi\rangle.
  $$
</Figure>

この定理のいくつかの帰結についてはレッスンの中で議論しますが、まずは、先ほどのシュミット分解の議論からこの定理がどのように導かれるかを見てみましょう。

仮定は、$\vert\psi\rangle$と$\vert\phi\rangle$が、$\mathsf{X}$の状態を表すある密度行列$\rho$について等式

$$
\operatorname{Tr}_{\mathsf{Y}} (\vert\psi\rangle\langle\psi\vert) = \rho =
\operatorname{Tr}_{\mathsf{Y}} (\vert\phi\rangle\langle\phi\vert)
$$

を満たす、系の組$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルであるということです。

$\rho$のスペクトル分解を考えます。

$$
\rho = \sum_{a = 0}^{n-1} p_a \vert x_a\rangle\langle x_a\vert
$$

ここで、$\{\vert x_0\rangle,\ldots,\vert x_{n-1}\rangle\}$は$\rho$の固有ベクトルからなる正規直交基底です。
先ほど説明した手順に従うと、$\vert\psi\rangle$と$\vert\phi\rangle$の両方について、次の形のシュミット分解が得られます。

$$
\begin{aligned}
\vert\psi\rangle & = \sum_{a = 0}^{r-1} \sqrt{p_a} \, \vert x_a\rangle \otimes \vert u_a\rangle\\[1mm]
\vert\phi\rangle & = \sum_{a = 0}^{r-1} \sqrt{p_a} \, \vert x_a\rangle \otimes \vert v_a\rangle
\end{aligned}
$$

これらの式で、$r$は$\rho$の階数であり、$\{\vert u_0\rangle,\ldots,\vert u_{r-1}\rangle\}$と$\{\vert v_0\rangle,\ldots,\vert v_{r-1}\rangle\}$は、$\mathsf{Y}$に対応する空間の正規直交なベクトルの集合です。

同じ空間にある要素数の等しい2つの正規直交集合については、1つ目の集合を2つ目の集合に変換するユニタリー行列が常に存在します。そこで、$a = 0,\ldots,r-1$について$U \vert u_a\rangle = \vert v_a\rangle$となるようにユニタリー行列$U$を選べます。
具体的には、このような行列$U$を求めるには、まずグラム・シュミットの直交化法を使って、正規直交集合を正規直交基底$\{\vert u_0\rangle,\ldots,\vert u_{m-1}\rangle\}$と$\{\vert v_0\rangle,\ldots,\vert v_{m-1}\rangle\}$に拡張します。ここで$m$は$\mathsf{Y}$に対応する空間の次元です。そして、次のようにします。

$$
U = \sum_{a = 0}^{m-1} \vert v_a\rangle\langle u_a\vert.
$$

すると、

$$
\begin{aligned}
(\mathbb{I}_{\mathsf{X}} \otimes U) \vert\psi\rangle
& = \sum_{a = 0}^{r-1} \sqrt{p_a} \, \vert x_a\rangle \otimes U \vert u_a\rangle\\
& = \sum_{a = 0}^{r-1} \sqrt{p_a} \, \vert x_a\rangle \otimes \vert v_a\rangle\\
& = \vert\phi\rangle,
\end{aligned}
$$

となり、証明が完了します。

以下に、純粋化のユニタリー同値性に関係する多くの興味深い例や帰結のうち、いくつかを紹介します。
レッスンの後半では、忠実度の文脈で、*ウルマンの定理*として知られる、もう1つのきわめて重要な帰結を見ます。

### 超高密度符号化

超高密度符号化のプロトコルでは、アリスとボブがe-bitを共有しています。つまり、アリスが量子ビット$\mathsf{A}$を、ボブが量子ビット$\mathsf{B}$を持ち、組$(\mathsf{A},\mathsf{B})$全体がベル状態$\vert\phi^{+}\rangle$にあります。
このプロトコルは、アリスが自分の量子ビット$\mathsf{A}$にユニタリー演算を施すことで、この共有状態を4つのベル状態$\vert\phi^+\rangle$、$\vert\phi^-\rangle$、$\vert\psi^+\rangle$、$\vert\psi^-\rangle$のいずれにでも変換できる方法を記述しています。
アリスはこれを行った後、$\mathsf{A}$をボブに送ります。そしてボブは組$(\mathsf{A},\mathsf{B})$を測定して、自分がどのベル状態を持っているかを確かめます。

4つのベル状態のどれについても、ボブの量子ビット$\mathsf{B}$の縮約状態は完全混合状態です。

$$
\operatorname{Tr}_{\mathsf{A}}(\vert\phi^+\rangle\langle\phi^+\vert) =
\operatorname{Tr}_{\mathsf{A}}(\vert\phi^-\rangle\langle\phi^-\vert) =
\operatorname{Tr}_{\mathsf{A}}(\vert\psi^+\rangle\langle\psi^+\vert) =
\operatorname{Tr}_{\mathsf{A}}(\vert\psi^-\rangle\langle\psi^-\vert) =
\frac{\mathbb{I}}{2}
$$

純粋化のユニタリー同値性により、各ベル状態について、$\vert\phi^+\rangle$をそのベル状態に変換する、アリスの量子ビット$\mathsf{A}$だけに作用するユニタリー演算が存在しなければならないことがすぐにわかります。
これによってプロトコルの正確な詳細がわかるわけではありませんが、純粋化のユニタリー同値性から、超高密度符号化が可能であることがただちに導かれます。

また、ベル状態を完全混合状態の純粋化からなる任意の正規直交基底に置き換えれば、より大きな系への超高密度符号化の一般化も常に可能であることがわかります。

### 暗号への影響

純粋化のユニタリー同値性は、量子情報を使った暗号プリミティブの実装に関して影響を持ちます。
たとえば、純粋化のユニタリー同値性から、量子情報を使って理想的な形の*ビットコミットメント*を実装することは不可能であることがわかります。

ビットコミットメントのプリミティブには、（互いを信用していない）2人の参加者アリスとボブが関わり、2つのフェーズがあります。

* 1つ目のフェーズは*コミット*フェーズで、これを通じてアリスは2値の値$b\in\{0,1\}$にコミットします。
  このコミットメントは、*拘束的*（binding）、つまりアリスが考えを変えられないものでなければならず、また*秘匿的*（concealing）、つまりアリスがどちらの値にコミットしたかをボブが見分けられないものでなければなりません。
* 2つ目のフェーズは*公開*フェーズで、アリスがコミットしたビットがボブに知らされます。ボブは、公開された値が本当にコミットされた値であると確信できなければなりません。

直感的で操作的な言い方をすると、ビットコミットメントの1つ目のフェーズは、アリスが紙に2値の値を書き、その紙を金庫の中にしまって鍵をかけ、鍵は自分で持ったまま金庫をボブに渡すかのように機能すべきです。
金庫はボブの手元にあるので、アリスは紙に書いた2値の値にコミットしたことになります（したがって拘束的です）。しかし、ボブは金庫を開けられないので、アリスがどちらの値にコミットしたかはわかりません（したがって秘匿的です）。
2つ目のフェーズは、アリスが金庫の鍵をボブに渡し、ボブが金庫を開けてアリスがコミットした値を明らかにできるかのように機能すべきです。

実は、量子情報だけを使って完全なビットコミットメントのプロトコルを実装することは不可能です。それは純粋化のユニタリー同値性に矛盾するからです。
以下は、このことを示す議論の大まかな要約です。

まず、プロトコルの実行中、アリスとボブはユニタリー演算を行うか、新たに初期化された系を導入するかのどちらかしか行わないと仮定できます。
すべてのチャネルがスタインスプリング表現を持つという事実により、この仮定を置くことができます。

プロトコルのコミットフェーズの終わりには、ボブはある複合系を手元に持っており、それは2つの量子状態のどちらかにあるはずです。アリスが値$0$にコミットした場合は$\rho_0$、値$1$にコミットした場合は$\rho_1$です。
プロトコルが完全に秘匿的であるためには、ボブはこの2つの状態を区別できてはならないので、$\rho_0 = \rho_1$でなければなりません。
（そうでなければ、これらの状態を確率的に識別する測定が存在してしまいます。）

しかし、アリスとボブはユニタリー演算しか使っていないので、コミットフェーズの後、プロトコルに関わるすべての系をまとめた状態は純粋状態でなければなりません。
具体的には、アリスが$0$にコミットしたときのプロトコルに関わるすべての系の純粋状態を$\vert\psi_0\rangle$、アリスが$1$にコミットしたときのものを$\vert\psi_1\rangle$とします。
アリスとボブの（複合系かもしれない）系を$\mathsf{A}$と$\mathsf{B}$で表すと、

$$
\begin{aligned}
\rho_0 & = \operatorname{Tr}_{\mathsf{A}}(\vert\psi_0\rangle\langle\psi_0\vert)\\[1mm]
\rho_1 & = \operatorname{Tr}_{\mathsf{A}}(\vert\psi_1\rangle\langle\psi_1\vert).
\end{aligned}
$$

となります。

完全に秘匿的なプロトコルでは$\rho_0 = \rho_1$が要求されることから、$\vert\psi_0\rangle$と$\vert\psi_1\rangle$は同じ状態の純粋化であることがわかります。したがって、純粋化のユニタリー同値性により、

$$
(U\otimes\mathbb{I}_{\mathsf{B}})\vert\psi_0\rangle = \vert\psi_1\rangle.
$$

を満たす、$\mathsf{A}$だけに作用するユニタリー演算$U$が存在しなければなりません。

したがってアリスは、$\mathsf{A}$に$U$を施すことでコミットメントを$0$から$1$に、$U^{\dagger}$を施すことで$1$から$0$に、自由に変更できます。そのため、ここで考えている仮想的なプロトコルは、拘束的であるという条件をまったく満たしません。

### ヒューストン・ジョザ・ウッターズの定理

このレッスンのこの部分で議論する純粋化のユニタリー同値性の最後の帰結は、ヒューストン・ジョザ・ウッターズの定理として知られる次の定理です。
（実は、これはこの名前で知られる定理を少し簡略化した主張です。）

<Figure title="定理">
  ヒューストン・ジョザ・ウッターズ: $\mathsf{X}$と$\mathsf{Y}$を系とし、$\vert\phi\rangle$を組$(\mathsf{X},\mathsf{Y})$の量子状態ベクトルとします。
  また、$N$を任意の正の整数、$(p_0,\ldots,p_{N-1})$を確率ベクトルとし、$\vert\psi_0\rangle,\ldots,\vert\psi_{N-1}\rangle$を、次の式を満たす$\mathsf{X}$の状態を表す量子状態ベクトルとします。

  $$
  \operatorname{Tr}_{\mathsf{Y}}\bigl(\vert\phi\rangle\langle\phi\vert\bigr) = \sum_{a = 0}^{N-1} p_a \vert\psi_a\rangle\langle\psi_a\vert.
  $$

  このとき、$(\mathsf{X},\mathsf{Y})$が状態$\vert\phi\rangle$にあるときにこの測定を$\mathsf{Y}$に対して行うと次の2つの主張が成り立つような、$\mathsf{Y}$に対する（一般の）測定$\{P_0,\ldots,P_{N-1}\}$が存在します。

  1. 各測定結果$a\in\{0,\ldots,N-1\}$は確率$p_a$で現れる。
  2. 測定結果$a$が得られたという条件のもとで、$\mathsf{X}$の状態は$\vert\psi_a\rangle$になる。
</Figure>

直感的に言えば、この定理は次のことを述べています。2つの系の純粋状態がありさえすれば、1つ目の系の縮約状態を純粋状態の凸結合として捉える*どのような*見方についても、2つ目の系に対する測定によって、1つ目の系についてのその見方を実質的に現実のものにできる、ということです。
数$N$は、$\mathsf{X}$や$\mathsf{Y}$の古典状態の数によって制限されるとは限らないことに注意してください。
たとえば、$\mathsf{X}$と$\mathsf{Y}$が量子ビットである一方で、$N = 1,000,000$ということもありえます。

この定理を、純粋化のユニタリー同値性を使って証明します。まず、古典状態集合が$\{0,\ldots,N-1\}$である新しい系$\mathsf{Z}$を導入します。
3つの系の組$(\mathsf{X},\mathsf{Y},\mathsf{Z})$の、次の2つの量子状態ベクトルを考えます。

$$
\begin{aligned}
\vert\gamma_0\rangle & = \vert\phi\rangle_{\mathsf{XY}}\otimes\vert 0\rangle_{\mathsf{Z}}\\[1mm]
\vert\gamma_1\rangle & = \sum_{a = 0}^{N-1} \sqrt{p_a}\, \vert\psi_a\rangle_{\mathsf{X}} \otimes \vert 0\rangle_{\mathsf{Y}} \otimes \vert a\rangle_{\mathsf{Z}}
\end{aligned}
$$

1つ目のベクトル$\vert\gamma_0\rangle$は、与えられた量子状態ベクトル$\vert\phi\rangle$と、新しい系$\mathsf{Z}$の$\vert 0\rangle$とのテンソル積にすぎません。
2つ目のベクトル$\vert\gamma_1\rangle$は、本質的には、定理を自明にしてしまう量子状態ベクトルです（少なくとも、$\mathsf{Y}$を$\mathsf{Z}$に置き換えたとすれば）。$\mathsf{Z}$に対して標準基底測定を行うと、各結果$a$が確率$p_a$で得られ、その結果が得られたという条件のもとで$\mathsf{X}$の状態が$\vert\psi_a\rangle$になることは明らかだからです。

組$(\mathsf{Y},\mathsf{Z})$を、トレースアウトすると$\mathsf{X}$が残る1つの複合系と考えると、状態

$$
\rho = \sum_{a = 0}^{N-1} p_a \vert\psi_a\rangle\langle\psi_a\vert.
$$

の異なる2つの純粋化が得られたことがわかります。

具体的には、1つ目については

$$
\operatorname{Tr}_{\mathsf{YZ}} (\vert\gamma_0\rangle\langle\gamma_0\vert)
= \operatorname{Tr}_{\mathsf{Y}} (\vert\phi\rangle\langle\phi\vert) = \rho
$$

であり、2つ目については

$$
\begin{aligned}
\operatorname{Tr}_{\mathsf{YZ}} (\vert\gamma_1\rangle\langle\gamma_1\vert)
& = \sum_{a,b = 0}^{N-1} \sqrt{p_a}\sqrt{p_b} \, \vert\psi_a\rangle\langle\psi_a\vert
\operatorname{Tr}(\vert 0\rangle\langle 0\vert \otimes \vert a\rangle\langle b\vert)\\
& = \sum_{a = 0}^{N-1} p_a \, \vert\psi_a\rangle\langle\psi_a\vert\\
& = \rho.
\end{aligned}
$$

です。

したがって、純粋化のユニタリー同値性により、

$$
(\mathbb{I}_{\mathsf{X}} \otimes U) \vert \gamma_0 \rangle = \vert\gamma_1\rangle
$$

を満たす、$(\mathsf{Y},\mathsf{Z})$に対するユニタリー演算$U$が存在しなければなりません。

このユニタリー演算$U$を使うと、次の図に示すように、定理の要件を満たす測定を実装できます。
言葉で説明すると、$\vert 0\rangle$状態に初期化された新しい系$\mathsf{Z}$を導入し、$(\mathsf{Y},\mathsf{Z})$に$U$を施します。これによって$(\mathsf{X},\mathsf{Y},\mathsf{Z})$の状態は$\vert\gamma_0\rangle$から$\vert\gamma_1\rangle$に変換されます。その後、$\mathsf{Z}$を標準基底測定で測定します。すでに見たように、これによって望みどおりの振る舞いが得られます。

$$
P_a = (\mathbb{I}_{\mathsf{Y}} \otimes \langle 0\vert) U^{\dagger}
(\mathbb{I}_{\mathsf{Y}} \otimes \vert a\rangle\langle a \vert)U (\mathbb{I}_{\mathsf{Y}} \otimes
