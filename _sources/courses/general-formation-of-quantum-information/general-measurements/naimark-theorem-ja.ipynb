# ナイマルクの定理

*ナイマルクの定理*は、測定に関する基本的な事実です。
この定理は、任意の一般測定が、チャネルのスタインスプリング表現を思わせるようなシンプルな方法で実装できることを示しています。

1. まず、測定対象の系を初期化された作業用の系と組み合わせて、複合系を作ります。
2. 次に、その複合系に対してユニタリー演算を行います。
3. 最後に、作業用の系を*標準基底測定*によって*測定*し、元の一般測定の結果を得ます。

## 定理の主張と証明

$\mathsf{X}$をある系とし、$\{P_0,\ldots,P_{m-1}\}$を次の条件を満たす半正定値行列の集まりとします。

$$
P_0 + \cdots + P_{m-1} = \mathbb{I}_{\mathsf{X}},
$$

これは、これらの行列が$\mathsf{X}$の測定を記述していることを意味します。
さらに、$\mathsf{Y}$を、古典状態集合が$\{0,\ldots,m-1\}$である系とします。これは、この測定で得られる可能性のある結果の集合です。

ナイマルクの定理は、複合系$(\mathsf{Y},\mathsf{X})$に対するユニタリー演算$U$が存在して、次の図で示される実装が、与えられた測定$\{P_0,\ldots,P_{m-1}\}$と一致する測定結果を与えることを述べています。つまり、それぞれの測定結果が得られる確率が、元の測定と厳密に一致するということです。

![ナイマルクの定理における一般測定の実装](https://quantum.cloud.ibm.com/learning/images/courses/general-formulation-of-quantum-information/general-measurements/Naimark.svg)

念のため説明すると、系$\mathsf{X}$は何らかの任意の状態$\rho$から始まり、$\mathsf{Y}$は$\vert 0\rangle$状態に初期化されます。
ユニタリー演算$U$が$(\mathsf{Y},\mathsf{X})$に対して適用され、その後、系$\mathsf{Y}$が標準基底測定によって測定され、何らかの結果$a\in\{0,\ldots,m-1\}$が得られます。

系$\mathsf{X}$は回路の出力の一部として図示されていますが、ここでは、$U$が実行された後の$\mathsf{X}$の状態については特に考えず、トレースアウトされていると考えることにします。
ただし、$U$が実行された後の$\mathsf{X}$の状態については、このレッスンの後半で取り上げます。

このような方法による測定の実装は、チャネルのスタインスプリング表現を明らかに思わせるものであり、その数学的な基礎も同様です。
ここでの違いは、作業用の系がスタインスプリング表現の場合のようにトレースアウトされるのではなく、測定されるという点です。

すべての測定がこの方法で実装できるという事実は、証明するのはかなり簡単ですが、そのためにはまず、半正定値行列に関する事実が必要になります。

**"事実"**

  $P$を$n \times n$の半正定値行列とします。$Q^2 = P$を満たす$n\times n$の半正定値行列$Q$が一意に存在します。この一意な半正定値行列は$P$の*平方根*と呼ばれ、$\sqrt{P}$と表記されます。


半正定値行列の平方根を求める1つの方法は、まずスペクトル分解を計算することです。

$$
P = \sum_{k=0}^{n-1} \lambda_k \vert \psi_k \rangle \langle \psi_k \vert
$$

$P$が半正定値であるため、その固有値は非負の実数でなければならず、それらを平方根に置き換えることで、$P$の平方根の表現を得ることができます。

$$
\sqrt{P} = \sum_{k=0}^{n-1} \sqrt{\lambda_k} \vert \psi_k \rangle \langle \psi_k \vert
$$

この概念を手にしたので、ナイマルクの定理を証明する準備が整いました。
$\mathsf{X}$が$n$個の古典状態を持つと仮定すると、対$(\mathsf{Y},\mathsf{X})$に対するユニタリー演算$U$は、$nm\times nm$の行列として表すことができ、これを、ブロックが$n\times n$である$m\times m$のブロック行列として見ることができます。
証明の要点は、$U$として、次のパターンに一致する任意のユニタリー行列を取ることです。

$$
U =
\begin{pmatrix}
\sqrt{P_0} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\sqrt{P_1} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\vdots & \vdots & \ddots & \vdots\\[1mm]
\sqrt{P_{m-1}} & \fbox{?} & \cdots & \fbox{?}
\end{pmatrix}
$$

疑問符が付いたブロックに何らかの値を埋め込んで$U$をユニタリーにすることが可能であるためには、ブロック$\sqrt{P_0},\ldots,\sqrt{P_{m-1}}$で構成される最初の$n$列が正規直交していることが必要十分です。
そうすれば、前のレッスンで扱ったのと同様に、グラム・シュミットの直交化法を使って残りの列を埋めることができます。

$U$の最初の$n$列は、次のようにベクトルとして表すことができます。ここで、$c = 0,\ldots,n-1$は、0から数えた列番号を表します。

$$
\vert\gamma_c\rangle = \sum_{a = 0}^{m-1} \vert a \rangle \otimes \sqrt{P_a} \vert c\rangle
$$

これらのうち任意の2つの間の内積は、次のように計算できます。

$$
\langle \gamma_c \vert \gamma_d \rangle =
\sum_{a,b = 0}^{m-1} \langle a \vert b \rangle \cdot \langle c \vert \sqrt{P_a}\sqrt{P_b}\, \vert d\rangle
= \langle c \vert \Biggl(\sum_{a = 0}^{m-1}  P_a \Biggr) \vert d\rangle
= \langle c \vert d\rangle
$$

これにより、これらの列が実際に正規直交していることが示されるので、行列全体がユニタリーになることを保証する形で、$U$の残りの列を埋めることができます。

残っているのは、シミュレーションの測定結果の確率が元の測定と一致することを確認することです。
$\mathsf{X}$の初期状態$\rho$が与えられたとき、集まり$\{P_0,\ldots,P_{m-1}\}$によって記述される測定は、それぞれの結果$a\in\{0,\ldots,m-1\}$を確率$\operatorname{Tr}(P_a \rho)$で生じさせます。

シミュレーションの結果の確率を求めるために、まず、$U$が実行された後の$(\mathsf{Y},\mathsf{X})$の状態に$\sigma$という名前を付けることにします。
この状態は、次のように表すことができます。

$$
\sigma =
U \bigl(\vert 0\rangle \langle 0 \vert \otimes \rho\bigr) U^{\dagger}
= \sum_{a,b=0}^{m-1} \vert a\rangle \langle b \vert \otimes \sqrt{P_a} \rho \sqrt{P_b}
$$

同じことをブロック行列の形式で表すと、次の式が得られます。

$$
\begin{aligned}
\sigma & =
\begin{pmatrix}
\sqrt{P_0} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\sqrt{P_1} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\vdots & \vdots & \ddots & \vdots\\[1mm]
\sqrt{P_{m-1}} & \fbox{?} & \cdots & \fbox{?}
\end{pmatrix}
\begin{pmatrix}
\rho & 0 & \cdots & 0 \\[1mm]
0 & 0 & \cdots & 0 \\[1mm]
\vdots & \vdots & \ddots & \vdots\\[1mm]
0 & 0 & \cdots & 0
\end{pmatrix}
\begin{pmatrix}
\sqrt{P_0} & \sqrt{P_1} & \cdots & \sqrt{P_{m-1}} \\[1mm]
\fbox{?} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\vdots & \vdots & \ddots & \vdots\\[1mm]
\fbox{?} & \fbox{?} & \cdots & \fbox{?}
\end{pmatrix}\\[5mm]
& = \begin{pmatrix}
\sqrt{P_0}\rho\sqrt{P_0} & \cdots & \sqrt{P_0}\rho\sqrt{P_{m-1}} \\[1mm]
\vdots & \ddots & \vdots\\[1mm]
\sqrt{P_{m-1}}\rho\sqrt{P_0} & \cdots & \sqrt{P_{m-1}}\rho\sqrt{P_{m-1}}
\end{pmatrix}
\end{aligned}
$$

疑問符が付いたブロックに属する$U$の成分は、結果には何の影響も与えないことに注意してください。これは、$\vert 0 \rangle \langle 0 \vert \otimes \rho$という形の行列を共役変換していることによるものです。
つまり、行列の積を計算する際、疑問符が付いた成分は常に$\vert 0 \rangle \langle 0 \vert \otimes \rho$のゼロの成分と掛け合わされることになります。

それでは、$\mathsf{Y}$に対して標準基底測定を行った場合に何が起こるかを分析してみましょう。
得られる可能性のある結果の確率は、$\mathsf{Y}$の縮約状態$\sigma_{\mathsf{Y}}$の対角成分によって与えられます。

$$
\sigma_{\mathsf{Y}} = \sum_{a,b=0}^{m-1} \operatorname{Tr}\Bigl(\sqrt{P_a} \rho \sqrt{P_b}\Bigr) \vert a\rangle \langle b \vert
$$

特に、トレースの巡回性を使うと、ある結果$a\in\{0,\ldots,m-1\}$が得られる確率が次のように求まることが分かります。

$$
\langle a \vert \sigma_{\mathsf{Y}} \vert a \rangle
= \operatorname{Tr}\Bigl(\sqrt{P_a} \rho \sqrt{P_a}\Bigr)
= \operatorname{Tr}(P_a \rho)
$$

これは元の測定と一致しており、シミュレーションの正しさが確認されました。

## 非破壊測定

このレッスンではこれまで、出力が古典的な測定結果のみで構成され、測定された系の測定後の量子状態については特に指定されない*破壊的*測定を扱ってきました。

一方、*非破壊*測定は、まさにこれを行うものです。
具体的には、非破壊測定は、古典的な測定結果の確率だけでなく、それぞれの測定結果が得られたという条件のもとでの、測定された系の状態も記述します。
*非破壊*という用語は、測定される*系*に対して使われるものであり、必ずしもその*状態*が非破壊であることを意味するわけではないことに注意してください。状態は、測定の結果として大きく変化する可能性があります。

一般に、ある破壊的測定が与えられたとき、その破壊的測定と*互換性のある*非破壊測定、つまり、古典的な測定結果の確率が破壊的測定と厳密に一致する非破壊測定は、複数(実際には無限に多数)存在します。
したがって、与えられた測定に対して、系の測定後の量子状態を定義する一意な方法は存在しません。

実際には、非破壊測定をさらに一般化することも可能で、古典的な測定結果に加えて、必ずしも入力系と同じではない系の量子状態出力を生成するようにできます。

非破壊測定という概念は、興味深く有用な抽象化です。
しかし、非破壊測定は常にチャネルと破壊的測定の合成として記述できるということは認識しておくべきです。つまり、ある意味では、破壊的測定という概念のほうがより基本的であると言えます。

### ナイマルクの定理から

ナイマルクの定理にあるような、一般測定のシミュレーションを考えてみましょう。
このシミュレーションから非破壊測定を得る簡単な方法は、先ほどの図で示されています。そこでは、系$\mathsf{X}$はトレースアウトされず、出力の一部となっています。
これにより、古典的な測定結果$a\in\{0,\ldots,m-1\}$と、$\mathsf{X}$の測定後の量子状態の両方が得られます。

これらの状態を数学的な言葉で記述してみましょう。
$\mathsf{X}$の初期状態は$\rho$であると仮定しているので、初期化された系$\mathsf{Y}$が導入されて$U$が実行された後、$(\mathsf{Y},\mathsf{X})$は次の状態になります。

$$
\sigma =
U \bigl(\vert 0\rangle \langle 0 \vert \otimes \rho\bigr) U^{\dagger}
= \sum_{a,b=0}^{m-1} \vert a\rangle \langle b \vert \otimes \sqrt{P_a} \rho \sqrt{P_b}.
$$

それぞれの古典的な結果が現れる確率は、先ほどと同じです。それらは、$\mathsf{X}$を無視するかどうかを私たちが決めることによって変わることはありません。
つまり、それぞれの$a\in\{0,\ldots,m-1\}$が確率$\operatorname{Tr}(P_a \rho)$で得られます。

特定の測定結果$a$が得られたという条件のもとで、$\mathsf{X}$の結果として得られる状態は、次の式で与えられます。

$$
\frac{\sqrt{P_a} \rho \sqrt{P_a}}{\operatorname{Tr}(P_a \rho)}
$$

これを理解する1つの方法は、$\mathsf{Y}$の標準基底測定を、完全脱位相化チャネル$\Delta_m$によって表すことです。ここで、チャネルの出力は、古典的な測定結果を(対角)密度行列として記述します。
得られる状態の表現は次のとおりです。

$$
\sum_{a,b=0}^{m-1} \Delta_m(\vert a\rangle \langle b \vert) \otimes \sqrt{P_a} \rho \sqrt{P_b}
= \sum_{a=0}^{m-1} \vert a\rangle \langle a \vert \otimes \sqrt{P_a} \rho \sqrt{P_a}.
$$

この状態は、次のように積状態の凸結合として表すことができます。

$$
\sum_{a=0}^{m-1} \operatorname{Tr}(P_a \rho)\, \vert a\rangle \langle a \vert \otimes \frac{\sqrt{P_a} \rho \sqrt{P_a}}{\operatorname{Tr}(P_a \rho)},
$$

これは、それぞれの測定結果が得られたという条件のもとでの$\mathsf{X}$の状態について、私たちが得た表現と一致しています。

### クラウス表現から

ナイマルクの定理の文脈における$U$の選び方には、同じ測定結果の確率を生じさせながらも、$\mathsf{X}$の出力状態が全く異なるものになる、別の選択肢があります。

例えば、1つの選択肢は、$U$を$(\mathbb{I}_{\mathsf{Y}} \otimes V) U$に置き換えることです。ここで、$V$は$\mathsf{X}$に対する任意のユニタリー演算です。
$\mathsf{X}$に対する$V$の適用は$\mathsf{Y}$の測定と交換可能なので、古典的な結果の確率は変わりませんが、結果$a$が得られたという条件のもとでの$\mathsf{X}$の状態は、次のようになります。

$$
\frac{V \sqrt{P_a} \rho \sqrt{P_a}V^{\dagger}}{\operatorname{Tr}(P_a \rho)}.
$$

さらに一般的には、$U$を次のユニタリー行列に置き換えることもできます。

$$
\Biggl(\sum_{a=0}^{m-1} \vert a\rangle\langle a \vert \otimes V_a\Biggr) U
$$

これは、$\mathsf{X}$に対する任意のユニタリー演算$V_0,\ldots,V_{m-1}$の選び方に対して成り立ちます。
この場合も、古典的な結果の確率は変わりませんが、結果$a$が得られたという条件のもとでの$\mathsf{X}$の状態は、次のようになります。

$$
\frac{V_a \sqrt{P_a} \rho \sqrt{P_a}V_a^{\dagger}}{\operatorname{Tr}(P_a \rho)}.
$$

この自由度を表現する同等の方法は、クラウス表現と結び付けられます。
つまり、$n$個の古典状態を持つ系の$m$個の結果を持つ非破壊測定は、クラウス行列に対する典型的な条件を満たす$n\times n$のクラウス行列$A_0,\ldots,A_{m-1}$の選び方によって記述できます。

$$
\sum_{a = 0}^{m-1} A_a^{\dagger} A_a = \mathbb{I}_{\mathsf{X}}
\tag{1}
$$

$\mathsf{X}$の初期状態が$\rho$であると仮定すると、古典的な測定結果が$a$になる確率は

$$
\operatorname{Tr}\bigl(A_a \rho A_a^{\dagger}\bigr)
= \operatorname{Tr}\bigl(A_a^{\dagger} A_a \rho \bigr)
$$

であり、結果が$a$であるという条件のもとで、$\mathsf{X}$の状態は次のようになります。

$$
\frac{A_a \rho A_a^{\dagger}}{\operatorname{Tr}(A_a^{\dagger}A_a \rho)}.
$$

これは、ナイマルクの定理におけるユニタリー演算$U$を次のように選ぶことと同等であることに注意してください。

$$
U =
\begin{pmatrix}
A_{0} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
A_{1} & \fbox{?} & \cdots & \fbox{?} \\[1mm]
\vdots & \vdots & \ddots & \vdots\\[1mm]
A_{m-1} & \fbox{?} & \cdots & \fbox{?}
\end{pmatrix}
$$

前のレッスンで、ブロック$A_0,\ldots,A_{m-1}$で構成される列が、条件$(1)$によって必然的に直交していることを確認しました。

### 一般化

非破壊測定を定式化する方法には、これまでに説明したものよりもさらに一般的な方法があります。
*量子インストルメント*(ここでは説明しません)という概念は、そのための1つの方法です。
