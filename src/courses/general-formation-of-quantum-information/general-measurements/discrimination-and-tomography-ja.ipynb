# 量子状態の識別とトモグラフィー

このレッスンの最後の部分では、測定に関連する2つの課題、*量子状態識別*と*量子状態トモグラフィー*について簡単に考えてみます。

1. *量子状態識別*

   量子状態識別では、既知の量子状態の集まり$\rho_0,\ldots,\rho_{m-1}$と、これらの状態に対応する確率$p_0,\ldots,p_{m-1}$が与えられます。
   これを簡潔に表現すると、量子状態の*アンサンブル*

   $$
   \{(p_0,\rho_0),\ldots,(p_{m-1},\rho_{m-1})\}
   $$

   が与えられているということになります。

   数$a\in\{0,\ldots,m-1\}$が確率$(p_0,\ldots,p_{m-1})$に従ってランダムに選ばれ、系$\mathsf{X}$が状態$\rho_a$に準備されます。
   目標は、$\mathsf{X}$のみを測定することによって、選ばれた$a$の値を決定することです。

   つまり、私たちには有限個の選択肢と、それぞれの$a$が選ばれる確率についての知識である*事前情報*があり、目標は実際にどの選択肢が起こったのかを決定することです。
   状態や確率の選び方によっては、これは簡単な場合もありますが、そうでない場合には、いくらかの誤りの可能性を伴わなければ決定できないこともあります。

2. *量子状態トモグラフィー*

   量子状態トモグラフィーでは、系の*未知*の量子状態が与えられます。つまり、量子状態識別とは異なり、通常は事前情報や、起こりうる選択肢についての情報はありません。

   しかし今回は、状態のコピーが1つだけ与えられるのではなく、多数の*独立な*コピーが与えられます。
   つまり、何らかの(場合によっては大きな)数$N$について、$N$個の同一の系$\mathsf{X}_1,\ldots,\mathsf{X}_N$がそれぞれ独立に状態$\rho$に準備されます。
   目標は、これらの系を測定することによって、未知の状態の近似を密度行列として求めることです。

## 2つの状態の識別

量子状態識別の最も単純な場合は、識別すべき状態が2つ、$\rho_0$と$\rho_1$である場合です。

ビット$a$がランダムに選ばれる状況を想像してみましょう。$a = 0$である確率が$p,$ $a = 1$である確率が$1 - p$です。
系$\mathsf{X}$は、$a$の値に応じて$\rho_0$または$\rho_1$である状態$\rho_a$に準備され、私たちに渡されます。
私たちの目標は、$\mathsf{X}$の測定によって$a$の値を正しく推測することです。
より正確に言えば、私たちの推測が正しい確率を最大化することを目指します。

### 最適な測定

この問題を解く最適な方法は、$\rho_0$と$\rho_1$の重み付き差(重みは対応する確率)のスペクトル分解から始まります。

$$
p \rho_0 - (1-p) \rho_1 = \sum_{k = 0}^{n-1} \lambda_k \vert \psi_k \rangle \langle \psi_k \vert
$$

この式では、プラス記号ではなくマイナス記号が使われていることに注意してください。これは重み付きの*和*ではなく、重み付きの*差*だからです。

次のようにして射影測定$\{\Pi_0,\Pi_1\}$を選ぶことで、正しく推測できる確率を最大化することができます。
まず、$\{0,\ldots,n-1\}$の要素を、重み付き差の対応する固有値が非負であるか負であるかに応じて、2つの互いに素な集合$S_0$と$S_1$に分割します。

$$
\begin{gathered}
S_0 = \{k\in\{0,\ldots,n-1\} : \lambda_k \geq 0 \}\\[2mm]
S_1 = \{k\in\{0,\ldots,n-1\} : \lambda_k < 0 \}
\end{gathered}
$$

これにより、次のように*射影*測定を選ぶことができます。

$$
\Pi_0 = \sum_{k \in S_0} \vert \psi_k \rangle \langle \psi_k \vert
\quad\text{および}\quad
\Pi_1 = \sum_{k \in S_1} \vert \psi_k \rangle \langle \psi_k \vert
$$

($\lambda_k = 0$となる$k$の値を、集合$S_0$と$S_1$のどちらに含めるかは実際には関係ありません。
ここでは任意に、これらの値を$S_0$に含めることにしています。)

これは、今考えている状況において、選ばれた状態の判定を誤る確率を最小化する最適な測定です。

### 正しさの確率

それでは、測定$\{\Pi_0,\Pi_1\}$が正しい結果を与える確率を求めてみましょう。

まず、$\Pi_0$と$\Pi_1$について私たちが行った具体的な選択について特に気にする必要はありませんが、それを念頭に置いておくと役立つかもしれません。
*任意の*測定$\{P_0,P_1\}$(必ずしも射影的でなくてもよい)について、正しさの確率は次のように書くことができます。

$$
p \operatorname{Tr}(P_0 \rho_0) + (1 - p) \operatorname{Tr}(P_1 \rho_1)
$$

$\{P_0,P_1\}$が測定であること、つまり$P_1 = \mathbb{I} - P_0$であることを使うと、この式は次のように書き換えられます。

$$
p \operatorname{Tr}(P_0 \rho_0) + (1 - p) \operatorname{Tr}((\mathbb{I} - P_0) \rho_1)\hspace*{3cm}\\[1mm]
\begin{aligned}
& = p \operatorname{Tr}(P_0 \rho_0) - (1 - p) \operatorname{Tr}(P_0 \rho_1) + (1-p) \operatorname{Tr}(\rho_1)\\[1mm]
& = \operatorname{Tr}\bigl( P_0 (p \rho_0 - (1-p)\rho_1) \bigr) + 1 - p
\end{aligned}
$$

一方、代わりに$P_0 = \mathbb{I} - P_1$という置き換えを行うこともできます。
これによって値は変わりませんが、別の表現が得られます。

$$
p \operatorname{Tr}((\mathbb{I} - P_1) \rho_0) + (1 - p) \operatorname{Tr}(P_1 \rho_1)\hspace*{3cm}\\[1mm]
\begin{aligned}
& = p \operatorname{Tr}(\rho_0) - p \operatorname{Tr}(P_1 \rho_0) + (1 - p) \operatorname{Tr}(P_1 \rho_1)\\[1mm]
& = p - \operatorname{Tr}\bigl( P_1 (p \rho_0 - (1-p)\rho_1) \bigr)
\end{aligned}
$$

この2つの式は同じ値を持つので、これらを平均することで、この値についてさらに別の表現を得ることができます。
(2つの式を平均するのは、結果として得られる式を簡単にするための工夫にすぎません。)

$$
\frac{1}{2} \bigl(\operatorname{Tr}\bigl( P_0 (p \rho_0 - (1-p)\rho_1) \bigr) + 1-p\bigr)
+ \frac{1}{2} \bigl(p - \operatorname{Tr}\bigl( P_1 (p \rho_0 - (1-p)\rho_1) \bigr)\bigr)\\
= \frac{1}{2} \operatorname{Tr}\bigl( (P_0-P_1) (p \rho_0 - (1-p)\rho_1)\bigr)  + \frac{1}{2}
$$

ここで、なぜ$P_0$と$P_1$としてそれぞれ(上で指定した)射影$\Pi_0$と$\Pi_1$を選ぶことが理にかなっているのかが分かります。というのも、そうすることで、最終的な式のトレースをできるだけ大きくすることができるからです。
具体的には、

$$
(\Pi_0-\Pi_1) (p \rho_0 - (1-p)\rho_1) = \sum_{k = 0}^{n-1} \vert\lambda_k\vert \cdot \vert \psi_k \rangle \langle \psi_k \vert.
$$

したがって、トレースを取ると、固有値の*絶対値*の和が得られます。これは、重み付き差の*トレースノルム*として知られているものに等しくなります。

$$
\operatorname{Tr}\bigl( (\Pi_0-\Pi_1) (p \rho_0 - (1-p)\rho_1)\bigr)
= \sum_{k = 0}^{n-1} \vert\lambda_k\vert = \bigl\| p \rho_0 - (1-p)\rho_1 \bigr\|_1
$$

したがって、それぞれ確率$p$と$1-p$で与えられる$\rho_0$と$\rho_1$を、測定$\{\Pi_0,\Pi_1\}$によって正しく識別できる確率は、次のとおりです。

$$
\frac{1}{2} + \frac{1}{2} \bigl\| p \rho_0 - (1-p)\rho_1 \bigr\|_1
$$

これが、それぞれ確率$p$と$1-p$で与えられる$\rho_0$と$\rho_1$を正しく識別するための最適な確率であるという事実は、一般に*ヘルストロム・ホレボの定理*(あるいは単に*ヘルストロムの定理*)と呼ばれています。

## 3つ以上の状態の識別

3つ以上の状態がある場合の量子状態識別については、最適な測定に関する既知の閉じた形の解はありませんが、この問題を半正定値計画法として定式化することは可能であり、これにより、コンピューターを使って最適な測定を効率的に数値的に近似することができます。

また、*ホレボ・ユエン・ケネディ・ラックス*条件として知られる条件によって、状態識別課題における与えられた測定の最適性を*検証*(あるいは*反証*)することも可能です。
具体的には、アンサンブル

$$
\{(p_0,\rho_0),\ldots,(p_{m-1},\rho_{m-1})\},
$$

によって定義される状態識別課題について、測定$\{P_0,\ldots,P_{m-1}\}$が最適であるためには、次の行列

$$
Q_a = \sum_{b = 0}^{m-1} p_b \rho_b P_b - p_a \rho_a
$$

が、すべての$a\in\{0,\ldots,m-1\}$について半正定値であることが必要十分です。

例えば、4つの正四面体状態$\vert\phi_0\rangle,\ldots,\vert\phi_3\rangle$のうちの1つが一様ランダムに選ばれる量子状態識別課題を考えてみましょう。
正四面体測定$\{P_0,P_1,P_2,P_3\}$は、次の確率で成功します。

$$
\frac{1}{4} \operatorname{Tr}(P_0 \vert\phi_0\rangle\langle \phi_0 \vert) +
\frac{1}{4} \operatorname{Tr}(P_1 \vert\phi_1\rangle\langle \phi_1 \vert) +
\frac{1}{4} \operatorname{Tr}(P_2 \vert\phi_2\rangle\langle \phi_2 \vert) +
\frac{1}{4} \operatorname{Tr}(P_3 \vert\phi_3\rangle\langle \phi_3 \vert)
= \frac{1}{2}.
$$

計算により

$$
Q_a = \frac{1}{4}(\mathbb{I} - \vert\phi_a\rangle\langle\phi_a\vert) \geq 0
$$

が$a = 0,1,2,3$について成り立つことが分かるので、ホレボ・ユエン・ケネディ・ラックス条件により、これは最適です。

## 量子状態トモグラフィー

最後に、*量子状態トモグラフィー*の問題について簡単に説明します。
この問題では、未知の量子状態$\rho$の独立なコピーが多数$N$個与えられ、目標は$\rho$の近似$\tilde{\rho}$を再構成することです。
念のため説明すると、これは、$\rho$にできるだけ近い密度行列$\tilde{\rho}$の古典的な記述を見つけたいということを意味します。

この設定は、次のような別の方法で説明することもできます。
未知の密度行列$\rho$が選ばれ、私たちには$N$個の量子系$\mathsf{X}_1,\ldots,\mathsf{X}_N$へのアクセスが与えられます。これらはそれぞれ*独立に*状態$\rho$に準備されています。
したがって、複合系$(\mathsf{X}_1,\ldots,\mathsf{X}_N)$の状態は、

$$
\rho^{\otimes N} = \rho \otimes \rho \otimes \cdots \otimes \rho \quad \text{($N$回)}
$$

となります。
目標は、系$\mathsf{X}_1,\ldots,\mathsf{X}_N$に対して測定を行い、それらの測定結果に基づいて、$\rho$を精度良く近似する密度行列$\tilde{\rho}$を計算することです。
これは非常に興味深い問題であることが分かっており、現在も研究が続けられています。

この問題に取り組むための戦略にはさまざまな種類が考えられます。
例えば、系$\mathsf{X}_1,\ldots,\mathsf{X}_N$のそれぞれを順番に個別に測定し、測定結果の列を生成するという戦略を想像できます。
どの測定を行うかについては、*適応的*な選択と*非適応的*な選択を含め、さまざまな具体的な選び方が考えられます。
つまり、特定の系に対してどの測定を行うかという選択が、それ以前の測定結果に依存する場合と依存しない場合があります。
測定結果の列に基づいて、状態$\rho$に対する推測$\tilde{\rho}$が導かれますが、これを行うための方法もさまざまです。

別の方法として、集まり全体に対して単一の*同時測定*を行うというアプローチもあります。この場合、$(\mathsf{X}_1,\ldots,\mathsf{X}_N)$を単一の系として考え、その出力が状態$\rho$に対する推測$\tilde{\rho}$となる単一の測定を選びます。
これにより、個々の系を個別に測定する場合よりも良い推定が得られる可能性がありますが、すべての系に対する同時測定は、実装がはるかに難しくなる可能性が高いです。

### パウリ測定を用いた量子ビットトモグラフィー

ここでは、$\rho$が量子ビットの密度行列であるという単純な場合の量子状態トモグラフィーについて考えてみます。
それぞれ独立に状態$\rho$にある量子ビット$\mathsf{X}_1,\ldots,\mathsf{X}_N$が与えられていると仮定し、目標は$\rho$に近い近似$\tilde{\rho}$を計算することです。

私たちの戦略は、$N$個の量子ビット$\mathsf{X}_1,\ldots,\mathsf{X}_N$を、3つのパウリ行列$\sigma_x,$ $\sigma_y,$ $\sigma_z$それぞれに対応する、おおよそ同じ大きさの3つの集まりに分けることです。
その後、それぞれの量子ビットを次のように独立に測定します。

1. $\sigma_x$に対応する集まりに含まれる量子ビットそれぞれについて、$\sigma_x$測定を行います。これは、量子ビットが基底$\{\vert + \rangle, \vert -\rangle\}$、つまり$\sigma_x$の固有ベクトルからなる正規直交基底に関して測定されることを意味し、対応する測定結果は、2つの固有ベクトルに対応する固有値です。状態$\vert + \rangle$には$+1,$ 状態$\vert -\rangle$には$-1$です。$\sigma_x$に対応する集まりのすべての状態について結果を平均することで、期待値

   $$
   \langle + \vert \rho \vert + \rangle - \langle - \vert \rho \vert - \rangle = \operatorname{Tr}(\sigma_x \rho).
   $$

   の近似が得られます。

2. $\sigma_y$に対応する集まりに含まれる量子ビットそれぞれについて、$\sigma_y$測定を行います。この測定は$\sigma_x$測定と似ていますが、測定基底が$\sigma_y$の固有ベクトルである$\{\vert\! +\!i \rangle, \vert\! -\!i \rangle\}$である点が異なります。$\sigma_y$に対応する集まりのすべての状態について結果を平均することで、期待値

   $$
   \langle +i \vert \rho \vert \!+\!i \rangle - \langle -i \vert \rho \vert \!-\!i \rangle = \operatorname{Tr}(\sigma_y \rho).
   $$

   の近似が得られます。

3. $\sigma_z$に対応する集まりに含まれる量子ビットそれぞれについて、$\sigma_z$測定を行います。今回の測定基底は、$\sigma_z$の固有ベクトルである標準基底$\{\vert 0\rangle, \vert 1 \rangle\}$です。$\sigma_z$に対応する集まりのすべての状態について結果を平均することで、期待値

   $$
   \langle 0 \vert \rho \vert 0 \rangle - \langle 1 \vert \rho \vert 1 \rangle = \operatorname{Tr}(\sigma_z \rho).
   $$

   の近似が得られます。

それぞれの集まりについて測定結果を平均することで、近似

$$
\alpha_x \approx \operatorname{Tr}(\sigma_x \rho),\;
\alpha_y \approx \operatorname{Tr}(\sigma_y \rho),\;
\alpha_z \approx \operatorname{Tr}(\sigma_z \rho)
$$

が得られたら、$\rho$を次のように近似できます。

$$
\tilde{\rho} = \frac{\mathbb{I} + \alpha_x \sigma_x + \alpha_y \sigma_y + \alpha_z \sigma_z}{2} \approx
\frac{\mathbb{I} + \operatorname{Tr}(\sigma_x \rho) \sigma_x + \operatorname{Tr}(\sigma_y \rho) \sigma_y + \operatorname{Tr}(\sigma_z \rho) \sigma_z}{2}
= \rho.
$$

$N$が無限大に近づく極限では、*大数の法則*により、この近似は真の密度行列$\rho$に確率収束します。また、(*ヘフディングの不等式*のような)よく知られた統計的な限界を使って、近似$\tilde{\rho}$が$\rho$からどの程度ずれる可能性があるかの確率を限界付けることができます。

ただし、重要なこととして認識しておくべきなのは、この方法で得られる行列$\tilde{\rho}$は密度行列にならない場合があるということです。
具体的には、トレースは常に$1$に等しくなりますが、半正定値にならない場合があります。
このような近似$\tilde{\rho}$を密度行列に「丸める」ための既知の戦略にはいくつかの種類があり、その1つは、スペクトル分解を計算し、負の固有値を$0$に置き換えて、(得られた行列をそのトレースで割ることで)再正規化するという方法です。

### 正四面体測定を用いた量子ビットトモグラフィー

量子ビットトモグラフィーを行うための別の選択肢は、すべての量子ビット$\mathsf{X}_1,\ldots,\mathsf{X}_N$を、先に説明した正四面体測定$\{P_0,P_1,P_2,P_3\}$を使って測定することです。
つまり、

$$
P_0 = \frac{\vert \phi_0 \rangle \langle \phi_0 \vert}{2}, \quad
P_1 = \frac{\vert \phi_1 \rangle \langle \phi_1 \vert}{2}, \quad
P_2 = \frac{\vert \phi_2 \rangle \langle \phi_2 \vert}{2}, \quad
P_3 = \frac{\vert \phi_3 \rangle \langle \phi_3 \vert}{2}
$$

であり、

$$
\begin{aligned}
\vert \phi_0 \rangle & = \vert 0 \rangle\\
\vert \phi_1 \rangle & = \frac{1}{\sqrt{3}} \vert 0 \rangle + \sqrt{\frac{2}{3}} \vert 1 \rangle\\
\vert \phi_2 \rangle & = \frac{1}{\sqrt{3}} \vert 0 \rangle + \sqrt{\frac{2}{3}} e^{2\pi i/3} \vert 1 \rangle\\
\vert \phi_3 \rangle & = \frac{1}{\sqrt{3}} \vert 0 \rangle + \sqrt{\frac{2}{3}} e^{-2\pi i/3} \vert 1 \rangle.
\end{aligned}
$$

です。
それぞれの結果が得られる回数を、$a\in\{0,1,2,3\}$ごとに$n_a$と表すことにすると、$n_0 + n_1 + n_2 + n_3 = N$となります。
これらの数と$N$の比は、それぞれの結果の確率の推定値を与えます。

$$
\frac{n_a}{N} \approx \operatorname{Tr}(P_a \rho).
$$

最後に、次の注目すべき式を利用します。

$$
\rho = \sum_{a=0}^3 \Bigl( 3 \operatorname{Tr}(P_a \rho) - \frac{1}{2}\Bigr) \vert \phi_a \rangle \langle \phi_a \vert.
$$

この式を証明するために、正四面体状態の内積の絶対値の2乗に関する次の等式を使うことができます。これは直接計算によって確認できます。

$$
\bigl\vert \langle \phi_a \vert \phi_b \rangle \bigr\vert^2 =
\begin{cases}
1 & a=b\\
\frac{1}{3} & a\neq b.
\end{cases}
$$

4つの行列

$$
\begin{aligned}
\vert\phi_0\rangle \langle \phi_0 \vert & = \begin{pmatrix} 1 & 0\\[2mm] 0 & 0\end{pmatrix}\\[3mm]
\vert\phi_1\rangle \langle \phi_1 \vert & = \begin{pmatrix} \frac{1}{3} & \frac{\sqrt{2}}{3}\\[2mm]
\frac{\sqrt{2}}{3} & \frac{2}{3}\end{pmatrix}\\[3mm]
\vert\phi_2\rangle \langle \phi_2 \vert & = \begin{pmatrix} \frac{1}{3} & \frac{\sqrt{2}}{3}e^{-2\pi i/3}\\[2mm]
\frac{\sqrt{2}}{3}e^{2\pi i/3} & \frac{2}{3}\end{pmatrix}\\[3mm]
\vert\phi_3\rangle \langle \phi_3 \vert & = \begin{pmatrix} \frac{1}{3} & \frac{\sqrt{2}}{3}e^{2\pi i/3}\\[2mm]
\frac{\sqrt{2}}{3}e^{-2\pi i/3} & \frac{2}{3}\end{pmatrix}
\end{aligned}
$$

は線形独立なので、$b = 0,1,2,3$について$\rho = \vert\phi_b\rangle\langle\phi_b\vert$の場合にこの式が成り立つことを示せば十分です。
具体的には、

$$
3 \operatorname{Tr}(P_a \vert\phi_b\rangle\langle\phi_b\vert) - \frac{1}{2}
= \frac{3}{2} \vert \langle \phi_a \vert \phi_b \rangle \vert^2 - \frac{1}{2}
= \begin{cases}
1 & a=b\\
0 & a\neq b
\end{cases}
$$

であり、したがって

$$
\sum_{a=0}^3 \biggl( 3 \operatorname{Tr}(P_a \vert\phi_b\rangle\langle\phi_b\vert) - \frac{\operatorname{Tr}(\vert\phi_b\rangle\langle\phi_b\vert)}{2}\biggr) \vert \phi_a \rangle \langle \phi_a \vert = \vert \phi_b\rangle\langle \phi_b \vert.
$$

となります。

これにより、$\rho$の近似が次のように得られます。

$$
\tilde{\rho} = \sum_{a=0}^3 \Bigl( \frac{3 n_a}{N} - \frac{1}{2}\Bigr) \vert \phi_a \rangle \langle \phi_a \vert.
$$

この近似は常にトレースが1に等しいエルミート行列になりますが、半正定値にならない場合があります。
この場合、パウリ測定を用いる戦略と同様に、この近似を密度行列に「丸める」必要があります。
