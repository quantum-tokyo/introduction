# はじめに

測定は、量子情報と古典情報の間のインターフェースを提供します。
量子状態にあるシステムに対して測定が行われると、古典情報が抽出され、その量子状態について何らかの情報が明らかになります。そして一般に、その過程で量子状態が変化したり破壊されたりします。
簡略化された量子情報の形式（*量子情報の基礎*コースで紹介されている形式）では、通常、*射影測定*、その中でも最も単純な種類の測定である*標準基底測定*にのみ注目します。
しかし、測定の概念は射影測定を超えて一般化することができます。

このレッスンでは、より一般的な測定について考察します。
一般的な測定を数学的に記述するいくつかの異なる方法について議論し、それらをこのコースでこれまでに議論してきた概念と結びつけます。

また、測定に関連する2つの概念、すなわち*量子状態識別*と*量子状態トモグラフィー*についても見ていきます。
量子状態識別とは、量子計算や暗号においてよく見られる状況を指し、システムが既知の状態の集合の中のいずれか1つに用意されており、測定によってどの状態が用意されたのかを判定することが目標となるものです。
一方、量子状態トモグラフィーでは、1つの未知の量子状態の独立なコピーが多数用意されており、それらのコピーに対して測定を行うことで、その状態の密度行列による記述を再構成することが目標となります。
