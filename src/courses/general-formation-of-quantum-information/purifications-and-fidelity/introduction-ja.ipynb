# はじめに

このレッスンでは、量子情報の理論における根本的に重要な概念である、状態の*純粋化*を中心に扱います。
密度行列$\rho$で表される量子状態の純粋化とは、より大きな複合系の純粋状態であって、複合系の残りの部分をトレースアウトすると$\rho$が残るもののことです。
これから見ていくように、トレースアウトされる部分が十分に大きければ、*すべての*状態$\rho$は純粋化を持ちます。

状態について推論する際に、その状態の純粋化を考えることはよく行われ、また有用でもあります。
直感的に言えば、量子状態ベクトルは密度行列よりも単純な数学的対象です。そのため、密度行列を、状態が純粋である（したがって、少なくともある点ではより単純な）より大きな系の一部を表すものとして考えることで、密度行列について興味深い結論を導けることがよくあります。
これは数学における*拡張*（dilation）の一例です。拡張とは、比較的複雑なものを、より大きいけれどもより単純なものを制限したり縮小したりすることで得ることを指します。

このレッスンでは、2つの量子状態の間の*忠実度*についても議論します。忠実度は、状態どうしの類似度を定量化する値です。
忠実度が数式によってどのように定義されるかを見て、それが*ウルマンの定理*を通じて純粋化の概念とどのように結びつくかを議論します。
