動画の内容を簡単に振り返り、補足します。

* ある問題が量子コンピューターで初めて解かれた後、古典スーパーコンピューターでそれを解く方法が見つかる、という事例がこれまでに見られています。古典コンピューティングと量子コンピューティングが互いを限界まで押し上げるこのサイクルは、今後数年間続くと考えられます。
* エラーの低減や利用可能な量子ビット数などの分野での進展を前提とすれば、量子コンピューティングが古典コンピューティングに対して証明可能な優位性を持ち得る特定の問題が存在します。しかし、現在はまだ探索の段階であり、量子に適したデータセットや有用な量子特徴マップを探し求めているところです。
* 量子機械学習 (QML) は、量子コンピューティングが既存の古典的なワークフローを拡張または補完できる、数多くの刺激的な分野の1つです。

機械学習 (ML) はデータセットにアルゴリズムを適用するものです。したがって QML では、データ側とアルゴリズム側のいずれか、あるいは両方に量子力学が含まれると考えるのが妥当でしょう。これらの可能性はいずれも興味深いものです。しかし本コースでは、主に古典データに適用される量子アルゴリズムの議論に限定します。その理由の1つは、古典データを用いた ML の問題はすでに十分に研究されており、広く利用可能であることです。古典データから出発する問題を解くことには幅広い関心が寄せられています。もう1つの理由は、QRAM が存在しないことです。大量の量子データを比較的長い時間スケールで保存する能力がない以上、量子データから出発する手法が産業に応用されるまでには、まだかなりの距離があります。また、古典データに効率的な方法で「量子的にアクセス」する方法も明らかではありません。特に関心の高い ML の種類として、ラベル付きデータセットを用いてアルゴリズムを訓練する教師あり学習と、ラベルのないサンプルから分布について学習しようとする教師なし学習の2つがあります。教師なしアルゴリズムは、例えば同じ分布から新しいサンプルを生成する方法や、類似した特徴を持つグループにサンプルをクラスタリングする方法を学習します。

左の画像は、教師あり学習におけるような、2つのカテゴリーのラベル付きデータを示しています。この場合、カテゴリーは線形分離可能です。右の画像はデータのクラスターを示しています。教師なし学習のタスクでは、これらのデータは最初はラベル付けされておらず、アルゴリズムが分布を調べ、例えばクラスターを探します。アルゴリズムが識別し得るクラスターの例を可視化するために、ここではデータ点にラベルを付けています。両者の重要な違いは、たとえ最終的にデータにラベルが付けられるとしても、教師あり学習のプロセスはすでにラベル付けされたデータから出発し、教師なし学習のプロセスはラベルのないデータから出発するという点です。

機械学習の素養がある方は、多くの解法がデータをより高次元の空間に写像することを含んでいることをすでにご存じでしょう。これは特にカーネルの文脈でよく研究されています。簡単に復習すると、データは、与えられたのと同じ次元数において、直線、平面、または超平面 (簡潔さのため、しばしば単に「超平面」と呼びます) によってカテゴリーに分離できる場合があります。これは上の1つ目の画像に示されています。一方、2つ目の画像に示すように、データがその次元では超平面によって分離できない場合もあります。しかしその場合でも、データには高次元への写像で活用できる構造が存在することがあり、写像後の高次元空間ではデータが分離可能になります。これは、円対称性を持つ2次元データを3次元空間に写像し、データ点が放物面上に並ぶ様子で示されています。

![QML\_CR\_background\_2D-3D.avif](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/introduction/qml-cr-background-2d-3d.avif)

QML における一般的な目標は、低次元の特徴の集合からより高次元の空間への写像を見つけ、データ点を効果的に分離し、その写像を用いて新しいデータ点を分類できるようにすることです。
しかしこれは容易なタスクではなく、機械学習における量子コンピューティングの潜在的な有用性を論じる際には、必ず適切な注意事項を伴わなければなりません。特に、データセットの選択における微妙な点と、ユーティリティー・スケールに到達する上での課題に取り組む必要があります。また、古典アルゴリズムによってすでに効率的かつ適切に扱われているデータで古典 ML アルゴリズムを上回ろうとすることから離れ、有用となり得る新しい特徴マップの探究へと議論の焦点を移す必要があります。

## 期待値の調整

文献で述べられている QML の応用に使われる多くのデータセットは「特徴量エンジニアリング」されたものです。つまり、量子コンピューティングが有用となる限定的なユースケースを示すために、特別に選択または生成されたデータセットです。これがずるいことのように思えるとしたら、それは取り組むべきタスクを誤解しています。ある量子特徴マップによって、すべての、あるいは多くの分類タスクを古典機械学習アルゴリズムよりも効率的に、またはスケーラブルに解けるようになるというわけでは**ありません**。そうではなく、一部の量子特徴マップ (すべてではありません) は古典的な特徴マップとは異なる振る舞いをするのです。したがって、取り組むべきタスクは、複雑なデータ構造の文脈で量子回路を探究することです。取り組むべき具体的な問いには、次のようなものがあります。

1. 古典的な代替手段と比べて、新しい振る舞いをする可能性が最も高い量子回路はどれか？
2. そのような新しい量子回路を用いて探究するのが最適な性質を持つデータを含む、現実世界の問題は存在するか？
3. これらの量子回路は近い将来の量子コンピューター上でスケールするか？

### 不十分な説明

量子コンピューティングがなぜ強力になり得るのかについて、単純化された説明をよく目にします。それは次のようなものです。

古典コンピューターが情報のビットを使うように、量子コンピューターは量子ビットを使います。ビット数を例えば4とすると、古典コンピューターは $2^4 = 16$ 通りの可能な状態のうちいずれか1つをとることができますが、量子コンピューターは16個すべての状態の重ね合わせとして同時に存在でき、この重ね合わせ全体に対して演算を実行できます。場合によっては、これにより、高次元空間への写像に基づく興味深い学習アルゴリズムを自然に設計できる可能性があります。

これは正しい記述ですが、後で説明するように不十分であり、やや誤解を招くものです。また、次のように複素係数と実係数の違いが強調されることもあります。

系が異なる状態にある確率をそれぞれ持つものとして記述できる確率的な古典系は、次のように記述できます。

$$
|s\rangle = a|0000\rangle+b|0001\rangle+c|0010\rangle+... a, b, c \in \reals
$$

このような系では、係数 $a$、$b$、$c$ などは正の実数である場合にのみ意味を持ちます。量子コンピューターにおける状態は、複素数となり得る確率振幅によって記述されます。

$$
|\psi \rangle = A|0000\rangle+B|0001\rangle+C|0010\rangle+... A, B, C \in \mathbb{C}
$$

上記の記述は、正しくなるよう非常に注意深く述べられています (表面上は似ていても誤っている記述は多数あります)。しかし、これらの正しい記述は、機械学習における量子コンピューティングの力を説明するものではありません。1つには、機械学習への量子コンピューティングの応用には必ず測定が伴い、量子ビットを測定して同時に複数の状態にあるという結果を得ることはできないからです。量子ビットを $|\psi\rangle = \frac{1}{\sqrt{2}}\left(|0\rangle+|1\rangle\right)$ のような重ね合わせ状態に準備することはできますが、測定すると $|0\rangle$ または $|1\rangle$ のいずれかが得られます。したがって、少なくとも、次元の増大に関するこの話は不完全です。さらに、カーネルの文脈では、ガウシアンカーネルは無限次元であるため、量子コンピューティングにおける次元の増大は、古典的な代替手段に対する計算能力の十分条件にはなり得ません。ここには微妙な点があり、ガウシアン特徴マップは、無限次元の写像ベクトルを計算する必要を回避する「カーネルトリック」と組み合わせてのみ使用されます。しかし、要点は変わりません。

**エンタングルした量子状態の高次元性は指数的な並列性ではなく、機械学習における能力向上の十分条件でもありません。**

以降のレッスンでは、量子回路を機械学習タスクに組み込むためのワークフローを紹介します。これは、量子コンピューティングの力の探究を促進するという明確な目的のために行うものです。本コースのどの特徴マップやアルゴリズムも、一般的な問題に対してより良い機械学習の結果を得るための近道として提示されるものではありません。なぜなら、そのような特徴マップやアルゴリズムは存在しないからです。そうではなく、有用な量子コンピューティングの探究に用いるための幅広い量子ツールを紹介します。

### 脱量子化

脱量子化 (dequantization) とは、ある量子アルゴリズムを、特定のタスク群に対して (通常はスケーリングも含めて) 量子アルゴリズムと同程度の性能を発揮する古典アルゴリズムで置き換えることを指します。定義によっては、古典アルゴリズムは量子アルゴリズムに比べて多項式的にしか遅くないことが求められます。

当初は古典アルゴリズムに対して大幅な高速化をもたらすと考えられていたいくつかの量子機械学習 (QML) アルゴリズムが、近年、脱量子化されています。この脱量子化のプロセスは、機械学習に対する量子的アプローチの潜在的な利点と限界について、重要な知見をもたらしました。

最も注目すべき脱量子化の成果の1つは、Ewin Tang による[推薦システムに関する研究](https://arxiv.org/abs/1807.04271)から生まれました。Tang は、それまで量子コンピューターでしか達成できないと考えられていた速度で推薦タスクを実行できる古典アルゴリズムを発見しました。この発見は、この問題に対して量子アルゴリズムが指数的な優位性を持つという前提に疑問を投げかけました。[Shin et al](https://journals.aps.org/prresearch/abstract/10.1103/PhysRevResearch.6.023218). によるより最近の研究では、変分量子機械学習モデルの関数クラスが脱量子化可能となる条件を特定することに焦点が当てられています。

脱量子化の一般的なアプローチの1つ (唯一の手法ではありません) は、データ読み込みのオーバーヘッドを考慮することです。つまり、古典データに適用される量子アルゴリズムには、必ず古典データを量子コンピューターにエンコードするステップがあります。量子アルゴリズムが、量子データがすでに利用可能な状態を出発点として仮定している場合、エンコードに必要な時間が事実上隠されてしまいます。量子データを仮定することが妥当な状況もありますが、関心のある多くの応用は古典データから出発します。いくつかの脱量子化の事例では、このエンコード時間を含め、かつ古典データの読み込みを効率的に行える場合には、量子アルゴリズムがもはや古典的な対応物を上回らないことが示されています。

あるアルゴリズムが脱量子化できないとしても、それがすべての古典アルゴリズムよりも効率的またはスケーラブルであることを意味するわけではありません。極端で人為的な例として、サイズ k の集合から最大の j 個の要素を選択するアルゴリズムを考えてみましょう。Shor のアルゴリズムを用いて k 個の各要素を素因数分解し、その素因数を用いて最大の要素を決定する量子アルゴリズムを書くことができます。このようなアルゴリズムはおそらく脱量子化できませんが、同じ最大要素の選択 (不要な素因数分解の部分は除く) を行う古典アルゴリズムと比べて、著しく非効率です。

## 存在証明

2021年、IBM Quantum® の研究者である Yunchao Liu、Srinivasan Arunachalam、Kristan Temme は、Nature に論文 [A rigorous and robust quantum speed-up in supervised machine learning](https://www.nature.com/articles/s41567-021-01287-z) を発表しました。上記の注意事項と整合するように、この研究では、(1) 古典的に困難であることが知られており、かつ (2) 量子アルゴリズムが高速化を示すのに適した分類問題が慎重に選ばれました。

この論文は、離散対数に基づくデータの分類を扱っています。論文を引用すると、「大きな素数 $p$ と $\mathbb{Z}^*_p = {1, 2, . . . , p − 1}$ の生成元 $g$ に対して、入力 $x \in \mathbb{Z}^*_p $ に対する $\text{log}_g(x)$ を、$p$ を表すのに必要なビット数 $n = \lceil{\text{log}_2(p)}\rceil$ の多項式時間で計算できる古典アルゴリズムは存在しない、というのが広く信じられている予想です。」これに対して、[Shor のアルゴリズム](https://epubs.siam.org/doi/10.1137/S0097539795293172)は離散対数問題を多項式時間で解けることが知られています。したがって、この問題の選択は、古典的な困難性 (脱量子化される可能性が低い) と、量子アルゴリズムに適していることが知られているという、上記の基準を同時に満たしています。

このように分類問題を賢明に選択することで、著者らは量子カーネル法 (以下で簡単に概説し、後のレッスンで詳しく説明します) を用いて、エンドツーエンドかつロバストな指数的高速化を示すことができました。ここで「エンドツーエンド」とは、古典データから出発するという仮定を指します。この場合、著者らはデータのエンコードにかかる時間も含めています。また「ロバスト」とは、分類されるデータが量子アルゴリズムによって大きなマージンで分離されるため、分類の成功が有限サンプリング誤差のような現実世界の要因に対してロバストであることを指します。

以上のことから、量子カーネルが指数的な高速化をもたらし得る問題は確かに存在すると言えます。しかし現在の科学の状況では、そのような問題は、量子アルゴリズムに適しているはずだという観察や理論的根拠に基づいて選ばれています。古典コンピューターがすでに十分にうまくこなしている機械学習タスクに対して、量子による高速化を期待するのは現実的ではありません。

量子ユーティリティーの探究のためにこのような理想的なケースを特定することは、本コースの受講者にとって非常に大きな責務です。そしてそれは、このようなコースの中で達成できるタスクではありません。その探究は、皆さんのような研究者で構成される IBM Quantum Network 全体にとっての課題です。本コースでは QML のワークフローとエンコード戦略を紹介し、皆さんが自身の専門分野で量子ユーティリティーの探究を始められるようにします。

この導入を通じて、量子機械学習についていくつかの点が明確になったことを願っています。

1. 量子アルゴリズムは、古典的に困難で、かつ量子アルゴリズムに適した、非常に特定の問題に対して、古典アルゴリズムに比べて指数的な高速化をもたらし得ます。
2. 量子コンピューティングにおけるエンタングルした状態の高次元性は重要ですが、それだけでは古典アルゴリズムに対する優位性を得るのに十分ではありません。
3. 量子アルゴリズムに適した問題を見つけることは極めて困難なタスクであり、その多くは本コースの受講者の皆さんに委ねられることになります。

## コースの学習目標

本コースを修了することで、以下の中核的なスキルと能力を身につけることが期待できます。受講者は次のことができるようになります。

1. QML とは何か、そして量子が古典機械学習とどこでつながるのかを説明する。

2. 量子の用語や重要な概念を ML のワークフローに適用する。

3. QML ワークフロー (さまざまな種類) の主要な構成要素を特定する。

4. さまざまな種類の QML を特定し、それらを区別する。

5. IBM Quantum プリミティブを使用し、Qiskit パターンに従って、量子カーネル法と変分量子分類器を実装する。

6. QML が最も有望な領域とそうでない領域を特定する。

7. 例題を自身のデータセットに合わせて調整する。

8. 訓練時間、ノイズ、複数状態の読み出しにおける誤差の累積など、QML における課題を認識する。

9. QML が自身の組織にどこで役立つかについて提言を行う。



## コースの構成

本コースはいくつかのレッスンで構成されています。各レッスンには本文中にいくつかの理解度チェックの問題があり、学習を進めながら新しいスキルを練習したり、理解度を確認したりできます。これらは必須ではありません。

コースの最後には20問のクイズがあります。Credly を通じて Quantum Machine Learning バッジを取得するには、このクイズで70%以上のスコアを獲得する必要があります。70%以上のスコアを獲得すると、その後まもなくバッジが自動的にメールで送信されます。クイズの提出は2回までです。1回目の提出後、間違えた問題に再挑戦する機会があります。2回目の提出後、スコアが確定します。詳細はクイズを参照してください。

コースの構成は以下のとおりです。

* レッスン 1: はじめにと概要
* レッスン 2: 機械学習の復習
* レッスン 3: データのエンコード
* レッスン 4: 量子カーネル法とサポートベクターマシン
* レッスン 5: 変分量子分類器 / ニューラルネットワーク

## 最初の QML コードを実行する

細部に分解したり背景を掘り下げたりする前に、どこに向かっているのかを見ておくと役に立つことがよくあります。以下のコードセルでは、量子カーネル法の簡単な例を実行します。具体的には、カーネル行列の要素を1つ計算します。カーネル法や量子カーネルに初めて触れる方も、これに尻込みする必要はありません。本コースの複数のレッスンを使って、これらのセルで行われている内容を詳しく解説していきます。

このコードと同時に、Qiskit パターンを紹介します。これは、ユーティリティー・スケールで量子コンピューティングに取り組むためのフレームワークです。このフレームワークは、非常に汎用的でほとんどの問題に適用できる4つのステップで構成されています (ただし、ワークストリームによっては、特定のステップが複数回繰り返される場合があります)。

### Qiskit パターン:

* ステップ 1: 古典的な入力を量子問題にマッピングする
* ステップ 2: 量子実行のために問題を最適化する
* ステップ 3: IBM Quantum プリミティブを使用して実行する
* ステップ 4: 解析 / 後処理

以下のセルでは、各ステップについてごく簡単な説明のみを示します。より詳しく学ぶために適切なレッスンを見つけられる程度の説明です。



```python
# 量子回路のセットアップに必要な qiskit パッケージをインポートします。
from qiskit.circuit import Parameter, ParameterVector, QuantumCircuit
from qiskit.circuit.library import unitary_overlap

# サンプラーとして StatevectorSampler をインポートします。
from qiskit.primitives import StatevectorSampler

# ステップ 1: 古典的な入力を量子問題にマッピングする:

# まず、適切なデータを取得します。
# 以下でインポートするデータは、128 行 (128 個のデータ点) で構成されています。
# 各行には、データの特徴に対応する 14 列と、ラベル (+/-1) を持つ 15 列目があります。
!wget https://raw.githubusercontent.com/qiskit-community/prototype-quantum-kernel-training/main/data/dataset_graph7.csv

# 必要なパッケージをインポートし、上で取得した csv ファイルから
# 訓練データを取り出す関数を作成します。
import pandas as pd
import numpy as np


def get_training_data():
    """訓練データを読み込みます。"""
    df = pd.read_csv("dataset_graph7.csv", sep=",", header=None)
    training_data = df.values[:20, :]
    ind = np.argsort(training_data[:, -1])
    X_train = training_data[ind][:, :-1]

    return X_train


# 訓練データを準備します
X_train = get_training_data()

# 空のカーネル行列
num_samples = np.shape(X_train)[0]

# 2 つのデータ点間の重なりを計算するための特徴マップを準備します。
# ZZFeatureMap のような既成の特徴マップを使うことも、ここで示すように
# カスタムの量子回路を使うこともできます。
num_features = np.shape(X_train)[1]
num_qubits = int(num_features / 2)
entangler_map = [[0, 2], [3, 4], [2, 5], [1, 4], [2, 3], [4, 6]]
fm = QuantumCircuit(num_qubits)
training_param = Parameter("θ")
feature_params = ParameterVector("x", num_qubits * 2)
fm.ry(training_param, fm.qubits)
for cz in entangler_map:
    fm.cz(cz[0], cz[1])
for i in range(num_qubits):
    fm.rz(-2 * feature_params[2 * i + 1], i)
    fm.rx(-2 * feature_params[2 * i], i)

# 2 つのデータ点 (ここでは 14 と 19) を選び、その特徴をパラメーターとして回路に割り当てます。
x1 = 14
x2 = 19
unitary1 = fm.assign_parameters(list(X_train[x1]) + [np.pi / 2])
unitary2 = fm.assign_parameters(list(X_train[x2]) + [np.pi / 2])

# 重なりを計算する回路を作成します
overlap_circ = unitary_overlap(unitary1, unitary2)
overlap_circ.measure_all()
overlap_circ.draw("mpl", scale=0.6, style="iqp")

# ステップ 2: 量子実行のために問題を最適化する

# 実際の量子コンピューターで実行するために、IBM Quantum Compute Service を使用して最も空いているバックエンドを取得します。
# from qiskit_ibm_runtime import QiskitRuntimeService

# service = QiskitRuntimeService(channel="ibm_quantum")
# backend = service.least_busy(
#    operational=True, simulator=False, min_num_qubits=overlap_circ.num_qubits
# )

# パスマネージャーを使用して、選択したバックエンド向けに回路を最適にトランスパイルします。
# from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
# pm = generate_preset_pass_manager(optimization_level=3, backend=backend)
# overlap_ibm = pm.run(overlap_circ)

# ステップ 3: IBM Quantum プリミティブを使用して実行する

# 使用するショット数を指定します。
num_shots = 10_000

## Qiskit の状態ベクトルベースのプリミティブを使用して問題を評価します
sampler = StatevectorSampler()
counts = (
    sampler.run([overlap_circ], shots=num_shots).result()[0].data.meas.get_int_counts()
)

# ステップ 4: 解析と後処理

# 0 が得られる確率を求めます。
counts.get(0, 0.0) / num_shots
```

上記のすべてのステップを理解する必要はありませんが、なぜこれを行っているのかを知るために、出力については理解しておきましょう。機械学習における多くのプロセスでは、(他の用途の中でも) 二値分類の一部として内積が用いられます。量子力学はこれと明らかなつながりがあります。なぜなら、さまざまな状態 $|\phi_i\rangle$ を測定する確率は、初期状態 $|\psi\rangle$ との内積によって $P_i = |\langle\phi_i|\psi\rangle|^2$ と与えられるからです。つまり、上で行ったのは、2つのデータ点の特徴を含み、それらを量子ベクトルの空間に写像する量子回路を作成し、測定を行うことでその空間における内積を推定することです。これは量子カーネル推定の一例です。このプロセスは2つのデータ点 (14番目と19番目) に対してのみ実行したことに注意してください。可能なすべてのペアに対してこれを行えば、出力 (この場合は数値 0.821...) を用いて、訓練データセット内のすべての点の間の重なりを記述する結果の行列を埋めることができます。これが「カーネル行列」です。
