# 量子機械学習

## 概要

このコースでは、量子コンピューティングが機械学習においてどのような刺激的な役割を果たすことができるのかを探ります。古典的な機械学習手法についても概説しますが、これらの手法に関するある程度の知識があることを前提としています。主な焦点は、量子コンピューティングを既存の機械学習ワークフローにどのように組み込むことができるかです。トピックには、データ符号化、量子カーネル、変分量子回路などが含まれます。
