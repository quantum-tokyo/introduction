# 関連する機械学習手法の復習

このセクションでは、量子機械学習のワークフローをより深く理解するのに役立つ、古典機械学習の重要な用語と手法を復習します。まず一般的な用語を紹介し、その後、2 種類の機械学習、すなわちカーネル法（特にサポートベクターマシンの文脈におけるもの）とニューラルネットワークについて詳しく見ていきます。これらの手法の間には確かに関連がありますが、ここおよび後のレッスンで議論する量子ワークフローの違いを踏まえ、両者を別個のものとして扱います。
これはあくまで概略であり、多くの細かな点は省略します。機械学習のより包括的な概要については、[\[1-3\]](#references) などの資料をお勧めします。


## 機械学習の種類

簡単に定義すると、機械学習とは、データ内のパターンや関係を分析し、そこから推論を導くアルゴリズムの集まりです。大まかに言えば、機械学習アルゴリズムは、扱うデータの種類と、明示的にプログラムされることなくアルゴリズムがどのように学習するかに応じて、主に 3 つのカテゴリーに分類できます。

1. **教師あり学習：** 教師あり学習では、モデルの学習に使用するデータにラベルが付けられています。これらのアルゴリズムの目的は、データとそれに対応するラベルまたは出力との関係を学習し、それを未知のデータに汎化することです。このクラスの代表的なタスクには、分類と回帰があります。
2. **教師なし学習：** 教師あり学習とは対照的に、教師なし学習ではラベルのないデータを使って機械学習モデルを学習させます。このようなアルゴリズムの目的は、データに潜むパターンや構造を発見することです。このクラスのアルゴリズムには、クラスタリングや次元削減のアルゴリズムがあります。敵対的生成ネットワークや変分オートエンコーダーなどの一部の生成モデルも、このカテゴリーに含めることができます。
3. **強化学習：** この機械学習カテゴリーのアルゴリズムは、環境と相互作用するエージェントによって定義されます。エージェントは行動を起こし、環境から報酬と罰という形でフィードバックを受け取ります。このフィードバックの仕組みを通じて、エージェントは最終的に、特定のタスクを遂行するための正しい一連の行動をとることを学習します。

![教師あり学習と教師なし学習の図式表現。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/qml-cr-background-sup-unsup.avif)

左の図は、教師あり学習のようにラベル付けされた 2 つのカテゴリーのデータを示しています。この場合、カテゴリーは線形分離可能です。右の図は、データのクラスターを示しています。教師なし学習のタスクでは、これらのデータには最初はラベルが付いておらず、アルゴリズムはその分布を調べ、たとえばクラスターを探します。アルゴリズムが識別しうるクラスターの例を可視化する目的で、ここではデータ点にラベルを付けています。両者の重要な違いは、たとえ最終的にデータにラベルが付けられるとしても、教師あり学習のプロセスはすでにラベル付けされたデータから始まり、教師なし学習のプロセスはラベルのないデータから始まるという点です。

### 機械学習への「量子」の導入

ここからは、「量子」が機械学習にどのように導入されるかを見ていきましょう。このより広い分類では、処理デバイス上のモデル／アルゴリズムの種類と、そこに与えられるデータの種類を考えます。上の図は、これらの可能な組み合わせをまとめたものです。

![一方の軸にアルゴリズム、もう一方の軸にデータをとり、それぞれが量子または古典である 4 つの象限を示す図。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/cccqqcqq-quadrants-cp.avif)

たとえば CC は、画像、音声、テキストなど古典コンピューターに保存できる古典的なデータセットがあり、機械学習アルゴリズムの実行にも古典コンピューターを使うことを意味します。これはまさに古典機械学習の設定です。一方、QQ は、量子コンピューターを使って量子データを処理することを意味します。ここでいう「量子データ」はいくつかの意味を持ちうるもので、文脈に依存することもあります。量子データは、量子デバイスから得られた測定結果の集合と考えることもできますし、別のアルゴリズムによって量子コンピューター上で準備された状態を指すこともあります。将来的には、現時点では存在しない QRAM（量子ランダムアクセスメモリー）に保存されたデータを指すことさえあるかもしれません。研究者が量子機械学習について語るとき、通常は CQ の領域、つまり手元のデータセットは古典的で、機械学習アルゴリズムを実行する処理デバイスが量子コンピューターである場合を指します。本コースの以降の部分では、このようなアルゴリズムに焦点を当てます。

## サポートベクターマシン

ここでは、サポートベクターマシンと呼ばれるアルゴリズムのクラスを、古典機械学習の観点から振り返ります。後ほど、このアルゴリズムに量子コンピューティングを取り入れる方法を示します。

![サポートベクターマシンの図。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/svm-cp.avif)

図に示すような 2 次元の特徴空間を持つデータセットに対して、二値分類のタスクを考えてみましょう。このデータセットを分類するためにできることの 1 つは、2 つのクラスを分離する直線、あるいは一般には超平面を見つけることです。実際には分離超平面は無数に見つかるため、問題は「最適なものをどう定義するか」です。ここでの考え方は、特に良い決定境界はマージンを最大化するべきだ、というものです。マージンは、各クラスの最も近い点までの距離として定義されます。この設定において、決定境界までの距離が最も小さいデータ点をサポートベクターと呼びます。

線形の決定境界はさまざまな方法で記述できますが、ある意味で最も素直な方法は、以下の $f_1$ に示すものです。ここで、$\Theta$ は超平面を定義するパラメーターの集合、$\vec{x}$ はデータセット、$b$ は定数のシフトです。$\Phi$ は入力データ点の空間からの写像であり、多くの場合（必ずしもそうではありませんが）より高次元の空間への写像です。この写像については後ほど改めて取り上げます。

モデル $f_1$ において、$\Theta$ はモデルが学習する調整可能なパラメーターのベクトルです。これを「主形式（primal formulation）」と呼びます。ある程度の数学的な操作により、同じ問題を定式化する 2 つ目の方法があることを示せます。これを「双対形式（dual formulation）」と呼び、以下の式 $f_2$ で表されます。この定式化では、アルファパラメーターについて最適化する必要があります。主な違いは、主形式では式に特徴ベクトルと学習可能なパラメーターとの内積が含まれるのに対し、双対形式では内積が特徴ベクトル同士の間でとられる点です。双対形式には学習データの特徴量と対応するラベルの両方が含まれますが、次のセクションで、双対形式が主形式よりもいかに有用であるかを見ていきます。

$$
f_1(\vec{x}) = \Theta^T \Phi(\vec{x})+b
$$

$$
f_2(\vec{x}) = \sum_{i=1}^n \alpha_i y_i \Phi^T(\vec{x}_i)\Phi(\vec{x})+b
$$

### 高次元空間への移行

この小節と次の小節では、より高次元への写像に焦点を当てて議論します。ここでのポイントは、空間間の写像という文脈で「カーネルトリック」を説明し、量子カーネルとは何かを理解するための準備をすることです。量子の波動関数の高い次元があらゆる問題を解決する、ということが要点では**ありません**。導入で述べたように、古典的なガウス特徴マップはすでに無限次元です。データの特徴量の次元は重要ですが、高次元の量子状態だけでは古典的手法に対する改善には十分ではありません。

図を見ると、適切な高次元への写像が与えられれば、元のデータが線形分離可能でない場合にも SVM のアプローチを一般化できることが容易にわかります。左の 2 次元データを見ると、2 つのクラスを分離できる線形の決定境界は存在しません。しかし、特徴空間に 3 つ目の特徴量を追加することを考えられます。この新しい特徴量が、たとえば元の 2 つの特徴量 $x_1$ と $x_2$ の原点からの距離であれば、データは線形分離可能になります。これは、このより高次元の特徴空間上であれば、サポートベクターマシンのアルゴリズムを問題なく実行できるようになることも意味します。

![一方のデータ種がリング状に分布し、そのリングの中央をもう一方のデータ種が埋めている様子を示す図。2 つ目のセルでは、データを 3 次元にお椀のような形で射影した様子を示しています。これでデータは線形分離可能になります。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/qml-cr-background-2d-3d.avif)

この「特徴マップ」も $\Phi$ で表します。特徴マップは、ここに示したように入力データの空間からより高い次元へ写像することが多いですが、より低い次元への写像を利用するモデルやアルゴリズムもあります。高次元への写像は、単に可視化しやすく理解しやすいケースだというだけです。

$$
\vec{x} = \begin{pmatrix}x_1 \\ x_2 \end{pmatrix}
$$

$$
\vec{\Phi}(\vec{x}) = \begin{pmatrix}x_1 \\ x_2 \\ {x_1}^2+{x_2}^2\end{pmatrix}
$$

特徴マップの中には、非常に高次元の空間へ写像するものもあります。そのような場合、高次元であるために内積の計算コストが高くなります。この点については後ほど改めて取り上げます。

### なぜ双対形式が有用なのか？

線形境界モデルの主形式と双対形式を思い出しましょう。

$$
f_1(\vec{x}) = \Theta^T \Phi(\vec{x})+b
$$

$$
f_2(\vec{x}) = \sum_{i=1}^n \alpha_i y_i \Phi^T(\vec{x}_i)\Phi(\vec{x})+b
$$

特徴マップを使ってより高次元の空間へ移ることで、分離超平面をうまく見つけられる場合があることがわかったので、式中の元の特徴ベクトル $\vec{x}$ を特徴マップ後のベクトルに置き換えることができます。しかし、これを主形式で行うと、パラメーターと、場合によっては非常に高次元となる特徴マップとの内積を計算しなければならないという問題に直面します。一方、双対形式では、これらが異なる入力の特徴マップ後のベクトル同士の内積に置き換わっていることがわかります。

特徴マップによっては、特徴マップ後のベクトルの内積 $\vec{\Phi}^T(\vec{x}_i)\cdot \vec{\Phi}(\vec{x}_j)$ を、元の（より低次元の）変数 $\vec{x}_i$ と $\vec{x}_j$ の簡単な関数 $k(\vec{x}_i, \vec{x}_j)$ として書ける場合があります。$\Phi$ の選び方によっては、$\vec{\Phi}^T(\vec{x}_i)\cdot \vec{\Phi}(\vec{x}_j)$ を低次元の内積 $\vec{x}_i^T\vec{x}_j$ の簡単な関数として書くことさえできるかもしれません。これは計算上非常に有利です。なぜなら、高次元での操作のコストをかけずに、データが線形分離可能となる空間にアクセスできるからです。実際、特徴マップ後のベクトルは $f_2$ の中で内積としてしか現れないため、内積を計算するために特徴マップを明示的に実行する必要さえないかもしれません。内積を計算する関数 $k(\vec{x}_i, \vec{x}_j)$ を「カーネル関数」と呼び、このように特徴マップの計算を回避する方法を「カーネルトリック」と呼びます。実際、特徴マップ後のベクトルが無限次元であっても、カーネルは非常に効率的に計算できる場合があります。

カーネル関数自体は、2 つの入力データベクトルの関数です。データセット内のデータベクトルの各ペアをカーネル関数の引数として代入すると、カーネル行列と呼ばれる対称な半正定値行列が得られます。

$$
k = \begin{pmatrix}k(\vec{x}_1,\vec{x}_1) & k(\vec{x}_1,\vec{x}_2) & ... \\ k(\vec{x}_2,\vec{x}_1) & k(\vec{x}_2,\vec{x}_2) & ... \\ \vdots & \vdots & \ddots\end{pmatrix}
$$

カーネル行列を計算したら、二次計画法のソフトウェアや「逐次最小最適化（sequential minimal optimization）」と呼ばれるアルゴリズムなどの手法を使って、最適なパラメーター（$\alpha_i$）を求めることができます。もちろん、これはデータのクラスを線形分離可能にする特徴マップに対応した、効率的に計算可能なカーネルが存在することを前提としています。これに関連しつつも新しいアプローチが、量子カーネル推定です。

### 量子カーネル

量子コンピューター、あるいはより一般に量子状態を用いると、「量子カーネル」を非常に自然に定義できます。入力 $\vec{x}$ を量子状態 $|\Phi(\vec{x})\rangle$ にエンコードすることを、特徴マップと解釈できます。このプロセスは、古典的な特徴マップでよく見られるように、実際にデータを非常に高次元の空間へ写像することがありますが、その次元はエンコード方法に依存します（「データエンコーディング」のレッスンを参照してください）。2 つの量子状態の内積 $\langle \psi | \phi \rangle$ は、状態 $|\psi\rangle$ にあるときに状態 $|\phi\rangle$ が測定される確率に関係していることを思い出してください。得られた回路を十分な回数測定することで、写像された 2 つのデータ点 $\vec{\Phi}(\vec{x}_i)$ と $\vec{\Phi}(\vec{x}_j)$ の内積を推定できます。

![カーネルを回路として抽象的に表現した図。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/classical-review-background-kernel-circuit.avif)

本コースの後半で見るように、上に示したような量子回路での測定を用いてカーネルを推定し、その後カーネル行列に対して古典的に SVM の最適化を実行して、調整可能なパラメーターを学習することができます。


## 変分量子分類器とニューラルネットワーク

もう 1 つの近い将来の量子機械学習アルゴリズムは、「変分量子回路（variational quantum circuits, VQCs）」と呼ばれます。これらの回路が分類タスクに用いられる場合、同じ略語が「変分量子分類器（variational quantum classifiers）」（これも VQCs）を指すのに使われることがあります。これらは古典的なニューラルネットワーク（NNs）と似た構造を利用することが多く、その場合は量子ニューラルネットワーク（QNNs）と呼ばれます。VQC はより一般的なものであり、必ずしも NN の構造に従う必要はないことを理解しておくことが重要ですが、既存の機械学習ワークフローにおいて量子が果たしうる役割を明確にするために、まずは NN とのアナロジーから始めます。その後、一般化について議論します。まずは古典的なニューラルネットワークを振り返りましょう。

以下の動画では、ニューラルネットワークと、それが変分量子回路と重なる部分について簡潔に復習します。詳しくは本文で掘り下げます。

<IBMVideo id="133981150" title="この動画では、Chris Porter がニューラルネットワークを簡潔に復習し、量子回路がどのように類似の構造を形成できるかを説明します。" />

ニューラルネットワークは、脳内のニューロンの構造と機能から大まかに着想を得た計算モデルです。図中のノードであるこれらのニューロンは層状に配置され、重みによって互いに接続されています。

![QML\_CR\_background\_QNN\_2ndlayer.png](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/qml-cr-background-qnn-2ndlayer.avif)

最初の層は入力層であり、この層のニューロンの活性 $a_n^0$ には、分析対象のデータ $\vec{x}$（たとえば画像の個々のピクセルの濃淡など）が直接入力されます。最後の層は出力層であり、分類結果を表します（画像の例で言えば、ある画像が犬である確率が 90%、猫である確率が 10% と分類する、といったものです）。

![QML\_CR\_background\_QNN\_output.png](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/qml-cr-background-qnn-output.avif)

各層のニューロンは、前の層から受け取った信号を処理し、それを重み $w_i$（図中の接続）を介して次の層へ伝えます。これらのニューロンの 1 つに着目すると、ニューラルネットワークの構成要素が得られます。これは「パーセプトロン」と呼ばれます。数学的には、パーセプトロンは入力ベクトル $\vec{x}$ を受け取り、学習可能な重みベクトルとの内積にバイアスを加えたものを計算します。そして非常に重要なこととして、パーセプトロンはこの計算結果に非線形の活性化関数（$\sigma$）を適用します。これらの非線形活性化関数は、ニューラルネットワークの高い表現力にとって不可欠です。別の見方をすると、もし層と層の間に非線形性がなければ、原理的にはニューラルネットワーク全体を 1 つの大きな行列積として書けてしまいます。その結果は単なる線形モデルとなり、深層ニューラルネットワークが捉えられるような複雑なパターンを捉えることはできません。したがって、非線形活性化関数はニューラルネットワークにおいて基本的な役割を担っています。

![perceptron\_CP.png](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/perceptron-cp.avif)

次のような関数

$$
f(\vec{x}) = \sigma (\vec{w}\cdot \vec{x}+\vec{b})
$$

が、既知のデータ $\vec{x}$ と非線形の $\sigma$、そして未知の重みベクトル $\vec{w}$ とバイアスベクトル $\vec{b}$ を用いて、各ニューロンで計算されます。一般に、すべての層のすべてのニューロン間に非ゼロの重みが存在しうるため、層 $L$ から層 $L+1$ へのニューロン $m$ と $n$ の間の重みを $w^L_{m,n}$ と表します。同様に、$L^\text{th}$ 層の $n^\text{th}$ ニューロンに対するバイアスを $b^L_n$ と表します。ここでのバイアスは、量子カーネルの議論に出てきた $b$ とは無関係です。

ニューラルネットワークは、重みとバイアスをランダムに設定した状態から始めることもあれば、既知の妥当な初期構成から始めることもあります。そこから、ニューラルネットワークがどの程度うまく分類できるかを確認し、改善していくというのが基本的な考え方です。ニューラルネットワークが正しい分類からどの程度ずれているかを表すために、コスト関数を用います。コスト関数の定義方法は数多くあります。ここでは、平均二乗誤差（MSE）を用いる一般的な例を 1 つ紹介します。

$$
C(w^L_{m,n},b^L_n) = \frac{1}{N}\sum_{i=1}^{N_\text{train}}\sum_{j=1}^{N_\text{outputs}}{(v_{i,j}-p_{i,j})^2}
$$

用途によって異なりますが、これは、学習データ中の画像 $i$ の出力 $j$ に対する実際の値 $v_{i,j}$（たとえば、出力層の「犬」のニューロンでは 1.0、その他すべてのニューロンでは 0 という値）と、予測値 $p_{i,j}$ との差をとることを意味します。その差を 2 乗してすべてのカテゴリーについて和をとることで、正しいカテゴリーが最も強く活性化されたかどうかだけでなく、誤った活性化が抑えられているかどうかも捉えられます。さらに、学習セット内のすべての例について和をとることで、コストが得られます。

次に、各層のすべてのニューロン間の重みや、すべてのニューロンのバイアスといったパラメーターを変化させます。コスト関数の局所的な最小値を探索するために、勾配降下法などの古典的な最適化ルーチンが用いられます。


### 量子パーセプトロン

パーセプトロンの量子版を構築するために考慮すべきことの 1 つは、古典ニューラルネットワークにおける活性化関数の役割である非線形性を、量子回路で実装できるようにすることです。というのも、特別な工夫をしない限り、量子回路はユニタリー演算、つまり単なる線形演算しか実装しないからです。量子回路に非線形性を導入するために使える方法はいくつかあります。主な方法の 1 つは、測定を非線形性の源として用いることです。その他の方法としては、量子フーリエ変換に基づく手法、回路途中の測定（mid-circuit measurement）や動的回路、回路から量子ビットをトレースアウトすることなどが挙げられます。

### 量子ニューラルネットワーク

量子ニューラルネットワーク（QNN）は、まず図中のユニタリー層 $U$ で入力データをエンコードし、次に層間の重みに対応する量子回路（下図の $W$）を適用し、最後に測定の層を置くことで動作します。これについての重要なポイントをいくつか挙げます。

* データの読み込みと重み付けは線形演算です。
* 測定は非線形です。
* したがって、古典 NN と同様に、線形な要素と非線形な要素の両方を備えています。
* 重みの回路には変分パラメーターが含まれるため、依然として古典的な最小化を行う必要があります。

![量子ニューラルネットワークを回路として表現した図。](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/qml-cr-background-qnn-ex-circuit.avif)

上のような回路を用いて、次の関数を計算できます。
$f_{QNN}(x) = \langle 0|U^{\dagger}(X)W^{\dagger}OWU(x)|0\rangle $
この関数は、一般に古典 NN で説明した関数 $f(x)$ と同じではないことに注意してください。特に、この関数には多数の重みからなる多数の層が含まれる可能性があり、$U$ によって量子回路に読み込まれたすべてのデータに対して適用されます。

### 一般化

ここで、ニューラルネットワークの量子版を構築する方法の 1 つを見てみましょう。このモデルでは、情報の流れが古典的なフィードフォワード型ニューラルネットワークとは異なります。古典的な設定では、情報は入力から始まりモデルの出力で終わるように左から右へ流れ、モデルを学習させるために誤差逆伝播法を行う際には逆方向に流れます。

![量子ニューラルネットワーク内の複数のゲート層を示す図](https://quantum.cloud.ibm.com/learning/images/courses/quantum-machine-learning/classical-ml-review/vqc-layers-meltem.avif)

しかし、この量子ニューラルネットワークの構成では、データをエンコードするユニタリーブロックが、学習可能なパラメーターを持つ変分ユニタリーブロックの間で繰り返されていることがわかります。「データ再アップロード（data reuploading）」と呼ばれるこの戦略は、興味深い理論的結果によって裏付けられています。実際、[Pérez-Salinas et al](https://quantum-journal.org/papers/q-2020-02-06-226/). の論文では、複数回のデータ再アップロードの助けを借りることで、「古典的なサブルーチンの支援があれば、単一の量子ビットでも普遍的な量子分類器を構築するのに十分な計算能力を持つ」ことが示されています。したがって、データ再アップロードは、モデルの表現力と表現能力を高めるために使える手法であり、量子ニューラルネットワークが複雑な関数を近似することを可能にします。
